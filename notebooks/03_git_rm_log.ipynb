# Git `rm` and `git log`

---

# Introduction

Two very important Git commands are:

- `git rm`
- `git log`

`git rm` is used for removing files from Git tracking.

`git log` is used for viewing commit history and navigating older versions of a project.

Understanding these commands is essential for safely managing files and project history.

---

# Part 1 — `git rm`

---

# What is `git rm`?

`git rm` removes files from:

- the working directory
- the staging area
- Git tracking

depending on the options used.

---

# Basic Syntax

```bash
git rm file.py
```

This:

- deletes the file from disk
- stages the deletion automatically

---

# Why use `git rm` instead of deleting manually?

Suppose you manually delete:

```bash
rm file.py
```

or delete it using File Explorer.

Git sees:

```text
deleted: file.py
```

but the deletion is NOT staged yet.

You still need:

```bash
git add .
```

or:

```bash
git rm file.py
```

---

# Benefit of `git rm`

`git rm` does BOTH operations together:

1. remove the file
2. stage the deletion

This is cleaner and safer.

---

# Manual delete vs `git rm`

| Method | Deletes file | Stages deletion |
|---|---|---|
| manual delete | Yes | No |
| `git rm` | Yes | Yes |

---

# Example

## Manual deletion

```bash
rm file.py
git status
```

Output:

```text
deleted: file.py
```

But deletion is unstaged.

You still need:

```bash
git add .
```

---

## Using `git rm`

```bash
git rm file.py
```

Now Git immediately stages the deletion.

---

# Q: What does:

```bash
git rm --cached file.py
```

do?

## Answer

This removes the file from Git tracking ONLY.

The actual file remains on disk.

This is extremely useful for:

- `.gitignore`
- removing large files from repository tracking
- keeping local files untracked

---

# Example

Suppose you accidentally committed:

```text
model.pt
```

Now you want:

- keep file locally
- stop tracking it in Git

Use:

```bash
git rm --cached model.pt
```

Then add it to `.gitignore`:

```text
model.pt
```

Then commit:

```bash
git commit -m "Stop tracking model.pt"
```

---

# Q: What is the difference between:

```bash
git rm file.py
```

and

```bash
git rm --cached file.py
```

## Answer

| Command | File Deleted From Disk | Removed From Git Tracking |
|---|---|---|
| `git rm file.py` | Yes | Yes |
| `git rm --cached file.py` | No | Yes |

---

# Q: What does `-r` mean?

## Answer

`-r` means:

> recursive

Used for folders/directories.

---

# Example

```bash
git rm -r my_folder/
```

This removes:

- the folder
- all files inside it
- stages all deletions

---

# Q: Why do we need `-r`?

## Answer

Git does not remove folders recursively by default.

Without `-r`:

```bash
git rm my_folder/
```

may fail.

---

# Q: What does `--force` mean?

## Answer

`--force` (or `-f`) forces Git to remove files even if:

- they have local modifications
- they are not fully committed

---

# Example

```bash
git rm -f file.py
```

---

# Why is `--force` dangerous?

Because it may destroy local work.

Git normally protects modified files.

`--force` overrides that protection.

---

# Q: When should we use `git rm`?

## Use `git rm` when:

- removing tracked files
- removing folders from Git
- removing files and staging deletion together
- stopping Git tracking with `--cached`

---

# Q: When should we avoid `git rm`?

## Avoid when:

- you only want local deletion temporarily
- file should remain tracked later
- you are unsure about losing modifications

---

# Common Examples

---

# Remove file completely

```bash
git rm file.py
```

---

# Remove folder recursively

```bash
git rm -r my_folder/
```

---

# Stop tracking but keep local file

```bash
git rm --cached model.pt
```

---

# Force removal

```bash
git rm -f file.py
```

---

# Part 2 — `git log`

---

# What is `git log`?

`git log` shows Git commit history.

It displays:

- commit hashes
- author
- date
- commit messages

---

# Basic Usage

```bash
git log
```

Example output:

```text
commit a1b2c3d
Author: Ali
Date: ...

    Added transformer tutorial

commit d4e5f6g
Author: Ali
Date: ...

    Fixed bug
```

---

# Q: What is a commit hash?

## Answer

A commit hash is a unique identifier for each commit.

Example:

```text
a1b2c3d
```

Git uses hashes to:

- identify commits
- restore old versions
- move between versions

---

# Q: What is:

```bash
git log --oneline
```

## Answer

It shows compact commit history.

Example:

```text
a1b2c3d Added transformer tutorial
d4e5f6g Fixed bug
```

Much shorter and easier to read.

---

# Why is `--oneline` useful?

Because normal `git log` can become very long.

`--oneline` gives a quick overview of project history.

---

# Q: How can we restore a file from an old commit using `git log`?

## Answer

Step 1 — Find commit hash:

```bash
git log -- file.py
```

Example:

```text
a1b2c3d Added file
d4e5f6g Modified file
```

---

Step 2 — Restore file:

```bash
git restore --source=a1b2c3d file.py
```

Now the file becomes the version from that commit.

---

# Q: How can we return the whole project to an older commit?

## Answer

Using `git reset`.

---

# Example

Suppose history:

```text
A -> B -> C -> D (HEAD)
```

You want to return to `B`.

Find hash:

```bash
git log --oneline
```

Suppose:

```text
1111111 Commit A
2222222 Commit B
3333333 Commit C
4444444 Commit D
```

Then:

```bash
git reset --hard 2222222
```

Now project becomes exactly like commit `B`.

---

# WARNING About `--hard`

This may destroy:

- local changes
- newer commits

Use carefully.

---

# Q: What is the difference between restoring a file and resetting the whole project?

## Answer

---

## Restore file

```bash
git restore --source=<hash> file.py
```

Only affects one file.

History remains unchanged.

---

## Reset project

```bash
git reset --hard <hash>
```

Moves HEAD and changes entire project history.

---

# Q: How can we inspect a specific commit?

## Answer

```bash
git show <commit-hash>
```

Example:

```bash
git show a1b2c3d
```

This shows:

- modified files
- inserted lines
- deleted lines
- commit message

---

# Q: How can we see logs in a graph style?

## Answer

```bash
git log --oneline --graph
```

Example:

```text
* a1b2c3d Added feature
* d4e5f6g Fixed bug
* h7i8j9k Initial commit
```

Very useful for branches.

---

# Useful `git log` Variants

| Command | Purpose |
|---|---|
| `git log` | full history |
| `git log --oneline` | compact history |
| `git log --graph` | branch graph |
| `git log -- file.py` | history of one file |
| `git show <hash>` | inspect one commit |

---

# Summary

---

# `git rm`

Used for:

- removing tracked files
- staging deletions automatically
- stopping tracking with `--cached`
- recursive folder deletion with `-r`

---

# `git log`

Used for:

- viewing project history
- finding commit hashes
- restoring old versions
- navigating Git history

---

# Important Safety Advice

Be careful with:

```bash
git reset --hard
git rm -f
```

because they may permanently remove work.
