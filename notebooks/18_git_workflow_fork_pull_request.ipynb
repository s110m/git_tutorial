# Workflow: Fork a Public Repository, Change It, and Send a Pull Request

This guide explains the full workflow:

```text
Public Repository
        ↓
Fork
        ↓
Clone Your Fork
        ↓
Create Branch
        ↓
Change Code
        ↓
Commit
        ↓
Push to Your Fork
        ↓
Open Pull Request to Original Repository
```

---

# 1. Big Picture

Suppose there is a public repository:

```text
https://github.com/original-owner/project.git
```

You want to:

```text
change some code
```

and send your changes to the original project.

But usually you do NOT have permission to push directly to the original repository.

So you use:

```text
fork + pull request
```

---

# 2. Visual Overview

```text
Original Repository
github.com/original-owner/project
        |
        | Fork
        v
Your Fork
github.com/your-username/project
        |
        | Clone
        v
Your Laptop
```

Then:

```text
Your Laptop
        |
        | Push
        v
Your Fork
        |
        | Pull Request
        v
Original Repository
```

---

# 3. Step 1 — Fork the Public Repository

Go to the original repository page on GitHub.

Example:

```text
github.com/original-owner/project
```

Click:

```text
Fork
```

GitHub creates your own copy:

```text
github.com/your-username/project
```

---

# What Fork Means

A fork is:

```text
your own GitHub copy of the original repository
```

You can change your fork freely.

The original repository is not changed.

---

# 4. Step 2 — Clone YOUR Fork

Now clone your fork, not the original repository.

```bash
git clone https://github.com/your-username/project.git
cd project
```

After cloning:

```text
Your Fork on GitHub
        |
        | clone
        v
Your Laptop
```

---

# 5. Step 3 — Check Remote Repositories

Run:

```bash
git remote -v
```

You may see:

```text
origin  https://github.com/your-username/project.git (fetch)
origin  https://github.com/your-username/project.git (push)
```

Here:

```text
origin = your fork
```

---

# 6. Step 4 — Add Original Repository as `upstream`

You should also connect your local project to the original repository.

```bash
git remote add upstream https://github.com/original-owner/project.git
```

Check again:

```bash
git remote -v
```

Now you should see:

```text
origin    https://github.com/your-username/project.git (fetch)
origin    https://github.com/your-username/project.git (push)

upstream  https://github.com/original-owner/project.git (fetch)
upstream  https://github.com/original-owner/project.git (push)
```

---

# Meaning of `origin` and `upstream`

| Name | Meaning |
|---|---|
| `origin` | Your fork |
| `upstream` | Original repository |

---

# Diagram

```text
Original Repository
        ^
        |
     upstream

Your Laptop

        |
        v
     origin
Your Fork
```

---

# 7. Step 5 — Update Your Local Main Branch

Before making changes, get the latest version from the original repository.

```bash
git switch main
git fetch upstream
git merge upstream/main
```

Or with rebase:

```bash
git switch main
git fetch upstream
git rebase upstream/main
```

Then update your fork:

```bash
git push origin main
```

---

# Why This Step Matters

The original repository may have changed after you forked it.

You want your work to start from the latest version.

---

# 8. Step 6 — Create a New Branch

Do not work directly on `main`.

Create a feature branch:

```bash
git switch -c fix-bug
```

Example branch names:

```text
fix-bug
update-readme
add-test
improve-docs
feature-login
```

---

# Diagram

Before:

```text
main:
A --- B --- C
```

After creating branch:

```text
main:
A --- B --- C

fix-bug:
A --- B --- C
```

After changes:

```text
main:
A --- B --- C

fix-bug:
A --- B --- C --- D
```

---

# 9. Step 7 — Change the Code

Edit files in VSCode or another editor.

Then check:

```bash
git status
```

You may see:

```text
modified: app.py
modified: README.md
```

---

# 10. Step 8 — See What You Changed

Before committing, inspect your changes:

```bash
git diff
```

This helps you avoid committing accidental changes.

---

# 11. Step 9 — Stage and Commit Changes

Stage files:

```bash
git add .
```

Or stage specific files:

```bash
git add app.py README.md
```

Commit:

```bash
git commit -m "Fix bug in data loading"
```

---

# Diagram

```text
main:
A --- B --- C

fix-bug:
A --- B --- C --- D
```

`D` is your commit.

---

# 12. Step 10 — Push Your Branch to Your Fork

Push your branch to your GitHub fork:

```bash
git push -u origin fix-bug
```

Remember:

```text
origin = your fork
```

So this uploads your branch to:

```text
github.com/your-username/project
```

not directly to the original repository.

---

# Diagram

```text
Your Laptop / fix-bug
        |
        | push
        v
Your Fork / fix-bug
```

---

# 13. Step 11 — Open Pull Request

Go to GitHub.

Usually GitHub shows:

```text
Compare & pull request
```

Click it.

Set:

```text
base repository: original-owner/project
base branch: main

head repository: your-username/project
compare branch: fix-bug
```

Meaning:

```text
Please merge my fix-bug branch from my fork
into the original repository main branch.
```

---

# Pull Request Direction

Very important:

```text
your fork / fix-bug  --->  original repo / main
```

Diagram:

```text
Your Fork: fix-bug
        |
        | Pull Request
        v
Original Repository: main
```

---

# 14. Step 12 — Write Pull Request Title and Description

Example:

```text
Title:
Fix bug in data loading

Description:
This PR fixes an issue where the data loader failed when the input folder was empty.

Changes:
- added empty-folder check
- improved error message
- added test case

How to test:
Run pytest tests/test_loader.py
```

---

# 15. Step 13 — What Happens After Pull Request?

The maintainers of the original repository can:

- review your code
- comment on lines
- request changes
- run tests
- approve the PR
- merge it
- close it

---

# 16. If Maintainer Requests Changes

You do NOT need to create a new pull request.

Use the same branch.

```bash
git switch fix-bug
```

Make changes.

Then:

```bash
git add .
git commit -m "Address review comments"
git push
```

The pull request updates automatically.

---

# Diagram

Before review update:

```text
fix-bug:
A --- B --- C --- D
```

After review update:

```text
fix-bug:
A --- B --- C --- D --- E
```

GitHub PR automatically includes commit `E`.

---

# 17. If Original Repository Changes Before PR Is Merged

Sometimes original `main` moves forward.

Example:

```text
Original main:
A --- B --- C --- F

Your branch:
A --- B --- C --- D
```

Now your branch may be behind.

---

# Update Your Branch

Option 1 — Merge:

```bash
git fetch upstream
git switch fix-bug
git merge upstream/main
```

If conflicts happen, resolve them.

Then:

```bash
git add .
git commit
git push
```

---

# Option 2 — Rebase:

```bash
git fetch upstream
git switch fix-bug
git rebase upstream/main
```

If conflicts happen:

```bash
git add .
git rebase --continue
```

Then push:

```bash
git push --force-with-lease
```

Use rebase carefully.

---

# 18. If There Is a Conflict

GitHub may say:

```text
This branch has conflicts that must be resolved
```

Resolve locally:

```bash
git fetch upstream
git switch fix-bug
git merge upstream/main
```

Open conflicted files.

You may see:

```text
<<<<<<< HEAD
your version
=======
upstream version
>>>>>>> upstream/main
```

Edit manually.

Then:

```bash
git add .
git commit
git push
```

---

# 19. After PR Is Merged

Your changes become part of original repository.

Then you can clean up.

---

# Update Local Main

```bash
git switch main
git fetch upstream
git merge upstream/main
git push origin main
```

---

# Delete Local Branch

```bash
git branch -d fix-bug
```

---

# Delete Remote Branch in Your Fork

```bash
git push origin --delete fix-bug
```

Or click:

```text
Delete branch
```

on GitHub.

---

# 20. Full Command Summary

```bash
# Clone your fork
git clone https://github.com/your-username/project.git
cd project

# Add original repository
git remote add upstream https://github.com/original-owner/project.git

# Update local main from original
git switch main
git fetch upstream
git merge upstream/main

# Update your fork main
git push origin main

# Create branch
git switch -c fix-bug

# Edit files...

# Check changes
git status
git diff

# Commit changes
git add .
git commit -m "Fix bug in data loading"

# Push branch to your fork
git push -u origin fix-bug
```

Then on GitHub:

```text
Open Pull Request:

from:
your-username/project:fix-bug

to:
original-owner/project:main
```

# 21. Common Mistakes

---

## Mistake 1 — Cloning Original Repository Instead of Fork

If you clone original repository and do not have write access:

```bash
git push
```

may fail.

Better:

```text
fork first, then clone your fork
```

---

## Mistake 2 — Working Directly on Main

Avoid working directly on:

```text
main
```

Use a branch:

```bash
git switch -c my-change
```

---

## Mistake 3 — Forgetting Upstream

Without upstream, it is harder to sync with original repository.

---

## Mistake 4 — Pull Request Direction Is Wrong

Correct:

```text
your fork branch -> original repository main
```

Wrong:

```text
original repository main -> your fork branch
```

---

# 22. Q & A

---

## Q1: Should I fork first or clone first?

Usually:

```text
fork first, then clone your fork
```

---

## Q2: Why not clone original repository directly?

You may not have permission to push to it.

---

## Q3: What is `origin`?

After cloning your fork:

```text
origin = your fork
```

---

## Q4: What is `upstream`?

Usually:

```text
upstream = original repository
```

---

## Q5: Where do I push my changes?

To your fork:

```bash
git push origin branch-name
```

---

## Q6: Does pushing to my fork change original repository?

No.

Original changes only after PR is merged.

---

## Q7: What is Pull Request here?

A request to merge your fork branch into the original repository.

---

## Q8: What if maintainers ask for changes?

Push more commits to the same branch.

The PR updates automatically.

---

## Q9: What if original repository changes while my PR is open?

Sync your branch with upstream using merge or rebase.

---

## Q10: Should I delete the branch after merge?

Usually yes, to keep repository clean.
