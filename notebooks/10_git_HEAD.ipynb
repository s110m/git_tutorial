# Understanding `HEAD` in Git

One of the most important concepts in Git is:

```text
HEAD
```

Many Git commands become much easier once you understand what `HEAD` really is.

---

# What is `HEAD`?

## Simple Definition

`HEAD` is:

> "Where you are currently located in Git."

It represents:

```text
your current checked-out position
```

---

# First Mental Model

Imagine Git history is a train track.

```text
A --- B --- C --- D
```

You are currently at commit `D`.

Then:

```text
HEAD -> D
```

means:

> "You are standing at commit D."

---

# More Accurate Definition

Usually:

```text
HEAD
```

does NOT point directly to a commit.

Instead:

```text
HEAD -> branch -> commit
```

Example:

```text
HEAD -> main -> D
```

Meaning:

- You are on branch `main`
- `main` currently points to commit `D`

---

# Visual Example

Suppose:

```text
A --- B --- C --- D   (main)
```

Normally:

```text
HEAD -> main -> D
```

Diagram:

```text
A --- B --- C --- D
                  ^
                main
                  ^
                HEAD
```

---

# Important Idea

## `HEAD` usually points to a BRANCH

NOT directly to commits.

---

# What Happens When You Commit?

Suppose:

```text
A --- B --- C   (main)
```

and:

```text
HEAD -> main
```

You create a new commit.

Result:

```text
A --- B --- C --- D   (main)
```

Now:

```text
HEAD -> main -> D
```

Git automatically moves:

- the branch
- and therefore HEAD

forward.

---

# Analogy

Imagine:

- Branch = bookmark in a book
- HEAD = your finger touching the bookmark

When bookmark moves, your finger moves too.

---

# Detached HEAD

Now suppose:

```bash
git switch --detach B
```

Result:

```text
A --- B --- C --- D   (main)
      ^
     HEAD
```

Now:

```text
HEAD -> B
```

directly.

NOT through a branch.

This is called:

```text
Detached HEAD
```

---

# Why Is It Called Detached?

Because HEAD is:

```text
detached from branches
```

Normally:

```text
HEAD -> main -> D
```

Detached:

```text
HEAD -> B
```

---

# What Happens If You Commit in Detached HEAD?

Suppose:

```text
A --- B --- C --- D   (main)
```

You detach:

```bash
git switch --detach B
```

Then commit:

```text
A --- B --- C --- D   (main)
      \
        E
```

Now:

```text
HEAD -> E
```

But:

```text
main -> D
```

Important:

- branch `main` does NOT know about `E`
- commit `E` may later become unreachable

---

# Why Detached HEAD Is Useful

Detached HEAD is useful for:

- testing old commits
- debugging old versions
- comparing behavior
- temporary experiments

without affecting branches.

---

# Relation Between HEAD and Branches

---

# Normal State

```text
HEAD -> branch -> commit
```

Example:

```text
HEAD -> main -> D
```

---

# Detached State

```text
HEAD -> commit
```

Example:

```text
HEAD -> B
```

---

# Relation Between HEAD and Working Directory

Your:

```text
working files
```

always reflect where HEAD points.

Example:

If:

```text
HEAD -> D
```

your files are contents of commit `D`.

If:

```bash
git switch --detach B
```

then:

```text
HEAD -> B
```

and your files become exactly like commit `B`.

---

# Relation Between HEAD and Staging Area

Git has 3 major areas:

```text
Working Directory
        ↓
Staging Area
        ↓
Repository (commits)
```

HEAD represents:

```text
current commit snapshot
```

Git often compares:

```text
Working Directory
vs
HEAD
```

to detect modifications.

---

# Example

Suppose HEAD commit contains:

```python
print("hello")
```

You modify file:

```python
print("new code")
```

Now:

```bash
git status
```

shows:

```text
modified file
```

because:

```text
Working Directory != HEAD
```

---

# Important Commands Related to HEAD

---

# 1. See Current HEAD

```bash
git log --oneline
```

Top commit is current HEAD commit.

---

# 2. Show HEAD Commit

```bash
git show HEAD
```

Shows:

- current commit
- diff
- metadata

---

# 3. Previous Commit

```bash
HEAD~1
```

means:

```text
one commit before HEAD
```

Example:

```text
A --- B --- C --- D
                  ^
                HEAD
```

Then:

```text
HEAD~1 = C
HEAD~2 = B
HEAD~3 = A
```

---

# 4. Compare with Previous Commit

```bash
git diff HEAD~1
```

Shows difference between:

```text
current working tree
and previous commit
```

---

# 5. Reset Relative to HEAD

Example:

```bash
git reset --hard HEAD~1
```

means:

```text
move branch one commit backward
```

---

# 6. Restore From HEAD

```bash
git restore app.py
```

means:

```text
restore app.py from HEAD version
```

---

# 7. Checkout/Switch HEAD

```bash
git switch main
```

changes:

```text
HEAD -> main
```

---

# HEAD vs Branch

| Concept | Meaning |
|---|---|
| Branch | Named pointer to commit |
| HEAD | Current location in Git |

---

# Example

```text
A --- B --- C --- D
                  ^
                main
                  ^
                HEAD
```

Here:

- `main` remembers latest commit of branch
- `HEAD` says where you currently are

---

# Special Syntax with HEAD

---

# HEAD^

Parent of HEAD:

```text
HEAD^
```

Usually same as:

```text
HEAD~1
```

---

# HEAD~n

Go back `n` commits.

Example:

```text
HEAD~3
```

means:

```text
3 commits before HEAD
```

---

# ORIG_HEAD

Git sometimes stores previous HEAD before dangerous operations.

Useful after:

```bash
git reset
```

or:

```bash
git merge
```

---

# Reflog and HEAD

Git tracks HEAD movements.

Use:

```bash
git reflog
```

Example output:

```text
HEAD@{0}: commit: added model
HEAD@{1}: checkout: moving from main to feature
HEAD@{2}: reset: moving to HEAD~1
```

This acts like Git's time machine.

---

# Real-World Workflow Examples

---

# Example 1 — Undo Last Commit

```bash
git reset --soft HEAD~1
```

Meaning:

```text
move branch one commit backward
```

---

# Example 2 — Compare with Previous Version

```bash
git diff HEAD~1
```

---

# Example 3 — Restore File from Current Commit

```bash
git restore app.py
```

Restores from:

```text
HEAD version
```

---

# Example 4 — Explore Old Commit

```bash
git switch --detach HEAD~3
```

---

# Q & A

---

## Q1: Is HEAD a commit?

Usually no.

Usually:

```text
HEAD -> branch -> commit
```

---

## Q2: Can HEAD point directly to commit?

Yes.

That is Detached HEAD state.

---

## Q3: Does HEAD move automatically?

Yes.

When you:

- commit
- switch branch
- checkout commit

HEAD changes.

---

## Q4: What does `HEAD~1` mean?

One commit before current HEAD.

---

## Q5: What does `HEAD^` mean?

Parent commit of HEAD.

Usually same as:

```text
HEAD~1
```

---

## Q6: Why is Detached HEAD dangerous?

Because new commits may become unreachable if no branch references them.

---

## Q7: Is Detached HEAD bad?

No.

Very useful for:

- testing
- debugging
- exploration

---

## Q8: How do I leave Detached HEAD?

Simply switch back:

```bash
git switch main
```

---

## Q9: Does HEAD affect working files?

Yes.

Your working directory reflects where HEAD points.

---

## Q10: What command shows HEAD history?

```bash
git reflog
```
