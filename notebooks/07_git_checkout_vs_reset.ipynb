# Git `checkout` vs `reset`

One of the most confusing concepts in Git is the difference between:

```bash
git checkout
```

and

```bash
git reset
```

The core idea is very simple:

---

# Main Difference

## `checkout`
Moves **YOU** to another commit.

## `reset`
Moves the **BRANCH** to another commit.

---

# Visual Example

Consider this history:

```text
A --- B --- C --- D   (main)
```

You are currently at commit `D`.

---

# 1. `git checkout B`

Command:

```bash
git checkout B
```

Result:

```text
A --- B --- C --- D   (main)
      ^
     HEAD
```

What happened?

- `main` still points to `D`
- Only `HEAD` moved to `B`
- You are temporarily visiting commit `B`
- Git enters **Detached HEAD** mode

You did NOT change history.

---

# Analogy: Reading an Old Page

Imagine Git history is a book.

- The branch (`main`) is a bookmark
- `checkout` means:

> "Go read page B temporarily."

The bookmark stays at page `D`.

You are just looking around.

---

# 2. `git reset --hard B`

Command:

```bash
git reset --hard B
```

Result:

```text
A --- B   (main)
```

What happened?

- The branch itself moved backward
- `C` and `D` disappeared from branch history
- You changed history

---

# Analogy: Moving the Bookmark

Now you are not just reading an old page.

You are moving the actual bookmark backward.

You are saying:

> "The branch officially ends at B now."

---

# Side-by-Side Comparison

| Feature | `checkout` | `reset` |
|---|---|---|
| Changes branch history? | No | Yes |
| Safe for exploration? | Yes | Sometimes dangerous |
| Moves HEAD? | Yes | Yes |
| Moves branch pointer? | No | Yes |
| Used for visiting old commits? | Yes | Not usually |
| Can remove commits? | No | Yes |

---

# Detached HEAD

After:

```bash
git checkout B
```

you are in:

```text
Detached HEAD
```

meaning:

```text
HEAD is directly pointing to a commit,
not to a branch.
```

Diagram:

```text
A --- B --- C --- D   (main)
      ^
     HEAD
```

Notice:

- HEAD is NOT attached to `main`
- `main` still points to `D`

---

# What Happens If You Commit After Checkout?

Suppose:

```text
A --- B --- C --- D   (main)
```

Then:

```bash
git checkout B
```

Then create new commits:

```text
A --- B --- C --- D   (main)
      \
        E --- F
```

You created a separate history path.

Unless you create a branch, those commits may later become unreachable.

---

# What Happens After Reset?

Suppose:

```text
A --- B --- C --- D   (main)
```

Then:

```bash
git reset --hard B
```

Result:

```text
A --- B   (main)
```

Now:

- `main` moved backward
- `C` and `D` are removed from branch history

---

# Different Types of Reset

---

## 1. Soft Reset

```bash
git reset --soft B
```

Effect:

```text
Commits removed
Changes kept staged
```

Analogy:

> "Undo commits, but keep work ready to recommit."

---

## 2. Mixed Reset (default)

```bash
git reset B
```

or

```bash
git reset --mixed B
```

Effect:

```text
Commits removed
Changes kept unstaged
```

---

## 3. Hard Reset

```bash
git reset --hard B
```

Effect:

```text
Commits removed
Changes deleted
Working directory overwritten
```

Dangerous.

---

# Safe Alternative: `git revert`

Instead of deleting history:

```bash
git revert C
```

Git creates a NEW commit that undoes `C`.

Before:

```text
A --- B --- C --- D
```

After revert:

```text
A --- B --- C --- D --- E
                    ^
                E undoes C
```

History is preserved.

---

# Analogy of Three Engineers

Imagine three engineers are editing the same machine design.

---

## Scenario 1: Checkout

Engineer Ali says:

> "Let me temporarily look at version B."

He studies old blueprints.

But the official company design remains unchanged.

That is:

```bash
git checkout B
```

---

## Scenario 2: Reset

Engineer Ali says:

> "Version B should officially become the latest design."

The company archive changes.

Versions `C` and `D` are removed from official history.

That is:

```bash
git reset --hard B
```

---

# Important Rule for Teams

## Usually Avoid:

```bash
git reset --hard
```

on shared branches.

Because it rewrites history.

---

## Prefer:

```bash
git revert
```

because it preserves history safely.

---

# Q & A

---

## Q1: Does `checkout` delete commits?

### No.

It only moves you temporarily to another commit.

---

## Q2: Does `reset` delete commits?

### Potentially yes.

Especially:

```bash
git reset --hard
```

---

## Q3: Why is `checkout` called detached HEAD?

Because HEAD points directly to a commit instead of a branch.

Normal:

```text
HEAD -> main -> D
```

Detached:

```text
HEAD -> B
main -> D
```

---

## Q4: Can I return after checkout?

Yes:

```bash
git checkout main
```

or:

```bash
git switch main
```

---

## Q5: Is `reset` dangerous?

`--hard` can be dangerous because it deletes local work.

---

## Q6: Can deleted commits be recovered?

Often yes.

Use:

```bash
git reflog
```

Git keeps a history of HEAD movements.

---

# Visual Summary

---

## Checkout

```text
A --- B --- C --- D   (main)
      ^
     HEAD
```

- Branch unchanged
- Temporary exploration

---

## Reset

```text
A --- B   (main/HEAD)
```

- Branch moved backward
- History rewritten

---

# Final Mental Model

## `checkout`

> "Take me to an old version temporarily."

---

## `reset`

> "Move the branch itself backward."

---

# Modern Git Alternatives

Git now prefers:

```bash
git switch
```

instead of checkout for branch switching.

And:

```bash
git restore
```

for restoring files.

Because `checkout` was doing too many unrelated things.
