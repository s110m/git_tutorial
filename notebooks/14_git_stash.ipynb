# Switching Branches Without Losing Unfinished Work

One very common situation in Git is this:

```text
You are working on something,
but it is NOT finished yet.
```

Then suddenly:

- you need to switch branch
- check another feature
- fix a bug
- pull latest updates
- test something

How can you switch branches safely without losing unfinished work?

This guide explains:

- how `git switch` behaves
- when Git blocks switching
- `commit` vs `stash`
- `git stash pop`
- `git stash apply`
- `git stash list`
- `git stash drop`
- and common workflows

---

# Big Picture

Suppose:

```text
main branch
feature branch
```

You are currently on:

```text
feature
```

and modified files:

```text
app.py
train.py
```

but did NOT commit yet.

Now you want:

```bash
git switch main
```

Sometimes Git allows this.

Sometimes Git blocks it.

Why?

---

# How `git switch` Thinks

Git tries VERY hard to:

```text
prevent accidental loss of work
```

---

# Scenario 1 — Safe Switch

Suppose:

You modified:

```text
app.py
```

and the target branch:

```text
main
```

does NOT have conflicting changes for that file.

Git may allow switching safely.

---

# Example

Current branch:

```text
feature
```

Modified:

```python
app.py
```

Target branch:

```text
main
```

has same base version for `app.py`.

Git can safely carry your modification.

---

# Result

```bash
git switch main
```

works.

Your unfinished changes move with you.

---

# Important Observation

YES, your understanding is mostly correct.

Git often allows switching when:

```text
your modifications do NOT conflict
with target branch files
```

---

# Scenario 2 — Dangerous Switch

Suppose:

Current branch:

```text
feature
```

You modified:

```python
app.py
```

But target branch:

```text
main
```

has DIFFERENT version of `app.py`.

Now switching could overwrite your work.

Git blocks it.

---

# Error Example

```text
error: Your local changes to the following files would be overwritten by checkout:
    app.py
Please commit your changes or stash them before you switch branches.
```

---

# Why Does Git Block It?

Because:

```text
switching would destroy your local modifications
```

Git protects you.

---

# Visual Diagram

Current branch:

```text
feature:
app.py = version A + your unfinished edits
```

Target branch:

```text
main:
app.py = completely different version
```

Git says:

> "I cannot safely combine these automatically."

---

# What Are Your Options?

Git suggests:

```text
1. commit
2. stash
```

---

# Option 1 — Commit

---

# What Is Commit?

```bash
git add .
git commit -m "unfinished work"
```

Commit means:

```text
save work permanently into git history
```

---

# Analogy

Commit is like:

> "Officially save this version into project history."

---

# Advantages of Commit

- permanent
- safe
- versioned
- visible in history
- easy to recover

---

# Disadvantages

Sometimes your work is:

- incomplete
- broken
- experimental
- temporary

You may not want permanent history yet.

---

# Option 2 — Stash

---

# What Is Stash?

```bash
git stash
```

means:

> "Temporarily hide my unfinished changes."

Git stores modifications in a hidden temporary area.

Then working directory becomes clean.

---

# Analogy

Imagine:

```text
unfinished papers on desk
```

`stash` means:

> "Put unfinished papers into drawer temporarily."

---

# After Stash

Before:

```text
modified files
```

After:

```text
clean working directory
```

Now switching branches is safe.

---

# Example Workflow

---

# Step 1 — Stash Work

```bash
git stash
```

---

# Step 2 — Switch Branch

```bash
git switch main
```

---

# Step 3 — Return Later

```bash
git switch feature
git stash pop
```

Your unfinished work returns.

---

# Commit vs Stash

| Feature | Commit | Stash |
|---|---|---|
| Permanent? | Yes | Temporary |
| Visible in history? | Yes | No |
| Good for unfinished work? | Sometimes | Yes |
| Shared with others? | Yes after push | No |
| Safe backup? | Very safe | Usually safe |

---

# Which One Is Preferred?

Depends on situation.

---

# Use Commit When

- work is meaningful
- work should be saved permanently
- work may be needed later
- team should see it

---

# Use Stash When

- temporary context switch
- quick branch change
- unfinished experiments
- short interruption

---

# Common Professional Advice

If work matters:

```text
prefer commit
```

because commits are safer and permanent.

Many developers create temporary commits like:

```bash
git commit -m "WIP"
```

(WIP = Work In Progress)

---

# Understanding `git stash pop`

Command:

```bash
git stash pop
```

means:

```text
restore latest stash
AND remove it from stash storage
```

---

# Why Is It Called "Pop"?

Like stack data structure in programming.

---

# Stack Analogy

Imagine stack of papers:

```text
Top
-----
stash 2
stash 1
stash 0
-----
Bottom
```

`pop` means:

```text
take top item out AND remove it
```

---

# What Happens Internally?

Before:

```text
stash list:
stash@{0}
stash@{1}
```

After:

```bash
git stash pop
```

Result:

```text
stash@{0} restored
stash@{0} removed from stash list
```

---

# Understanding `git stash apply`

Command:

```bash
git stash apply
```

means:

```text
restore stash
BUT keep stash stored
```

---

# Difference Between Pop and Apply

| Command | Restores Changes | Removes Stash |
|---|---|---|
| `git stash pop` | Yes | Yes |
| `git stash apply` | Yes | No |

---

# Analogy

---

# `pop`

```text
Take paper from drawer and remove it.
```

---

# `apply`

```text
Copy paper from drawer but keep original inside.
```

---

# When Use Apply?

Useful when:

- unsure if restoration will work
- want backup safety
- may reuse stash later

---

# Understanding `git stash list`

Command:

```bash
git stash list
```

Shows all stored stashes.

Example:

```text
stash@{0}: WIP on feature
stash@{1}: experiment changes
stash@{2}: debugging work
```

---

# Analogy

Imagine list of hidden saved unfinished works.

---

# Understanding Named Stashes

You can create stash with message:

```bash
git stash push -m "unfinished CNN model"
```

Now list:

```text
stash@{0}: unfinished CNN model
```

---

# Apply Specific Stash

Example:

```bash
git stash apply stash@{1}
```

or:

```bash
git stash pop stash@{1}
```

---

# Meaning

Restore specific stash instead of latest one.

---

# Example

Suppose stash list:

```text
stash@{0}: bug fix
stash@{1}: CNN experiment
stash@{2}: UI changes
```

Then:

```bash
git stash apply stash@{1}
```

restores:

```text
CNN experiment
```

without removing it.

---

# Understanding `git stash drop`

Command:

```bash
git stash drop stash@{1}
```

means:

```text
delete stash permanently
```

---

# Why Use Drop?

Sometimes stash list becomes large and messy.

You may no longer need old temporary works.

---

# Example

Before:

```text
stash@{0}
stash@{1}
stash@{2}
```

Drop:

```bash
git stash drop stash@{1}
```

After:

```text
stash@{0}
stash@{1}
```

(old stash@{2} shifts upward)

---

# Important Warning

Dropped stash may become difficult to recover.

---

# `git stash clear`

Dangerous command:

```bash
git stash clear
```

Deletes ALL stashes.

---

# Common Real Workflow

---

# Scenario

You are coding feature branch.

Suddenly production bug appears.

---

# Workflow

Save unfinished work:

```bash
git stash
```

Switch branch:

```bash
git switch main
```

Fix bug.

Commit and push.

Return:

```bash
git switch feature
git stash pop
```

Continue unfinished work.

---

# Can Stash Also Create Conflicts?

YES.

Suppose:

- stash modifies app.py
- current branch also changed app.py

Git may create merge conflicts during:

```bash
git stash pop
```

or:

```bash
git stash apply
```

---

# Conflict Resolution

Resolve manually.

Then:

```bash
git add .
```

# Q & A

---

## Q1: Why does Git sometimes block branch switching?

Because switching may overwrite local modifications.

---

## Q2: Does Git always block switching with modified files?

No.

Only when conflicts/danger may occur.

---

## Q3: Is stash permanent?

No.

Mostly temporary storage.

---

## Q4: Is commit safer than stash?

Usually yes.

Commits are permanent and versioned.

---

## Q5: What does `git stash pop` do?

Restore latest stash and remove it from stash list.

---

## Q6: What does `pop` mean?

Like stack data structure:

```text
take top item and remove it
```

---

## Q7: Difference between pop and apply?

`pop` removes stash after restore.

`apply` keeps stash.

---

## Q8: What is `git stash list`?

Shows all saved stashes.

---

## Q9: Can I apply specific stash?

Yes:

```bash
git stash apply stash@{1}
```

---

## Q10: What does `git stash drop` do?

Deletes stash permanently.

---

## Q11: Can stash restoration create conflicts?

Yes.

Especially if same files changed.
