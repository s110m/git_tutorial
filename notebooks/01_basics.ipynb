# Git Basics

---

# Introduction

Git is one of the most important tools in software engineering and machine learning projects.

It helps developers:

- track changes
- manage versions
- collaborate with teams
- restore older versions of code
- experiment safely using branches

Git is used almost everywhere in modern software development.

---

# Git vs GitHub

---

# Q: What is the difference between Git and GitHub?

## Answer

### Git

Git is:

- a version control system
- a tool for tracking code changes
- software installed on your computer

Git works locally.

---

### GitHub

GitHub is:

- a cloud platform for Git repositories
- a collaboration platform
- a remote storage for Git projects

GitHub uses Git internally.

---

# Q: Why do we need Git?

## Answer

Git solves many important problems.

---

## Problem 1 — Returning to older versions

Without Git, people often create folders like:

```text
project_final
project_final_v2
project_final_real
project_final_last
```

Git solves this professionally using commits.

---

## Problem 2 — Team collaboration

Git allows many developers to:

- work simultaneously
- edit different features
- merge changes safely

---

## Problem 3 — Experimentation

Git branches allow safe experimentation without damaging the main project.

---

# Commits

---

# Q: What is a commit?

## Answer

A commit is:

- a snapshot of the project
- a saved state of the repository
- a checkpoint in history

Commits form a chain:

```text
A -> B -> C
```

Each node is a commit.

---

# Q: Why are commits important?

## Answer

Commits allow us to:

- restore older versions
- understand project evolution
- debug problems
- collaborate effectively

---

# Branches

---

# Q: What is a branch?

## Answer

A branch is:

- an independent line of development
- a pointer to commits
- a separate workspace

---

# Example

Suppose we have:

```text
A -> B -> C (main)
```

Now create a branch:

```text
A -> B -> C
          \
           feature
```

Both branches initially point to commit `C`.

---

# Q: Why do we use branches?

## Answer

Branches allow:

- feature development
- experimentation
- bug fixing
- teamwork

without affecting the stable branch.

---

# Local vs Remote

---

# Q: What is a local repository?

## Answer

A local repository:

- exists on your computer
- stores commits locally
- contains the `.git` directory

---

# Q: What is a remote repository?

## Answer

A remote repository:

- exists on another server
- usually GitHub/GitLab
- is used for collaboration and backup

---

# Important Git Areas

Git can be viewed as three important areas:

```text
Working Directory  ->  Staging Area  ->  Repository
```

---

# 1. Working Directory

---

# Q: What is the working directory?

## Answer

The working directory contains:

- your actual files
- current modifications
- the code you edit

This is the project folder itself.

---

# Example

Suppose:

```python
print("hello")
```

You modify it:

```python
print("hello world")
```

This modification exists in the working directory.

---

# 2. Staging Area

---

# Q: What is the staging area?

## Answer

The staging area (index) is:

- a temporary preparation area
- where selected changes are prepared before commit

---

# Example

```bash
git add file.py
```

This stages the file.

---

# Q: Why do we need the staging area?

## Answer

The staging area allows:

- selective commits
- organized commits
- separating unrelated changes

---

# Example

Suppose you modified:

- `model.py`
- `README.md`
- `test.py`

You may only want:

```bash
git add model.py
git commit -m "Fix model bug"
```

without committing the other files.

---

# 3. Repository

---

# Q: What is the repository?

## Answer

The repository stores:

- commits
- project history
- snapshots of the project

---

# Git Status

---

# Q: What does `git status` do?

## Answer

```bash
git status
```

shows:

# Git Add

---

# Q: What does `git add` do?

## Answer

`git add` moves changes into the staging area.

---

# Example

```bash
git add file.py
```

Now the file is prepared for commit.

---

# Q: What does:

```bash
git add .
```

mean?

## Answer

It stages changes from the current directory.

Usually used for:

- modified files
- new files
- deleted files

inside the current folder.

---

# Q: What does:

```bash
git add --all
```

or

```bash
git add -A
```

mean?

## Answer

It stages:

- all modified files
- all new files
- all deleted files

throughout the repository.

---

# Q: What is the difference between:

```bash
git add .
```

and

```bash
git add -A
```

## Answer

| Command | Scope |
|---|---|
| `git add .` | current directory |
| `git add -A` | entire repository |

---

# Q: What does:

```bash
git add *
```

mean?

## Answer

`*` is expanded by the shell.

Example:

```bash
git add *
```

may become:

```bash
git add file1.py file2.py src/
```

---

# Why is `git add *` not recommended?

## Answer

Because:

- hidden files may be ignored
- behavior depends on shell
- less reliable

Usually prefer:

```bash
git add .
```

---

# Git Restore

---

# Q: What does:

```bash
git restore file.py
```

do?

## Answer

It restores the file from the current commit (`HEAD`).

Local modifications are discarded.

---

# Example

Suppose:

```python
print("new change")
```

was added accidentally.

Then:

```bash
git restore file.py
```

removes the modification.

---

# Q: What does:

```bash
git restore --staged file.py
```

do?

## Answer

It removes the file from staging.

The file modifications still remain.

---

# Q: What is the difference between:

```bash
git restore file.py
```

and

```bash
git restore --staged file.py
```

## Answer

| Command | Effect |
|---|---|
| `git restore file.py` | discard local modifications |
| `git restore --staged file.py` | unstage file |

---

# Q: Can `git restore` undo commits?

## Answer

No.

`git restore` mainly works with:

- files
- staging area
- working directory

It does NOT move Git history backward.

---

# Q: How can we restore a deleted file from older commits?

## Answer

Suppose:

```text
Commit A -> file exists
Commit B -> file deleted
```

Restore from older commit:

```bash
git restore --source=HEAD~1 file.py
```

---

# Q: Can we restore files from any old commit?

## Answer

Yes.

First find the commit:

```bash
git log -- file.py
```

Then:

```bash
git restore --source=<commit-hash> file.py
```

---

# Git Reset

---

# Q: What does `git reset` do?

## Answer

`git reset` mainly works with:

- commits
- HEAD
- branch history

It moves the current branch backward.

---

# Q: What does:

```bash
git reset HEAD~1
```

do?

## Answer

This removes the last commit but keeps file changes.

Equivalent to:

```bash
git reset --mixed HEAD~1
```

---

# Example

Before:

```text
A -> B -> C (HEAD)
```

After:

```text
A -> B (HEAD)
```

But changes from `C` still remain in files.

---

# Q: What does:

```bash
git reset --hard HEAD~1
```

do?

## Answer

This:

- removes last commit
- moves HEAD backward
- deletes file changes
- resets working directory

---

# Why is `--hard` dangerous?

## Answer

Because it may permanently destroy work.

---

# Q: What are the three reset modes?

## Answer

---

## 1. Soft reset

```bash
git reset --soft HEAD~1
```

Result:

- commit removed
- changes kept
- still staged

---

## 2. Mixed reset (default)

```bash
git reset HEAD~1
```

Result:

- commit removed
- changes kept
- unstaged

---

## 3. Hard reset

```bash
git reset --hard HEAD~1
```

Result:

- commit removed
- changes deleted

---

# Reset Comparison Table

| Command | Commit Removed | Changes Kept | Staged |
|---|---|---|---|
| `--soft` | Yes | Yes | Yes |
| `--mixed` | Yes | Yes | No |
| `--hard` | Yes | No | No |

# Git Log

---

# Q: What does `git log` do?

## Answer

`git log` shows commit history.

Example:

```bash
git log
```

---

# Example Output

```text
commit a1b2c3d
Author: Ali
Date: ...

    Added tutorial
```

---

# Q: What does:

```bash
git log --oneline
```

do?

## Answer

It shows compact history.

Example:

```text
a1b2c3d Added tutorial
f4e5d6g Fixed bug
```

---

# Q: Why is `git log --oneline` useful?

## Answer

Because normal logs may become very long.

`--oneline` gives a quick overview.

---

# Q: How can we inspect a commit?

## Answer

```bash
git show <commit-hash>
```

Example:

```bash
git show a1b2c3d
```

---

# Q: How can we return to an old version?

## Answer

Find commit:

```bash
git log --oneline
```

Then:

```bash
git reset --hard <commit-hash>
```

---

# WARNING

`--hard` may delete newer changes.

Use carefully.

---

# Git RM

---

# Q: What does `git rm` do?

## Answer

`git rm`:

- removes file from disk
- stages deletion automatically

---

# Example

```bash
git rm file.py
```

---

# Q: Why use `git rm` instead of manual deletion?

## Answer

Manual deletion:

```bash
rm file.py
```

does NOT stage deletion automatically.

`git rm` does both operations together.

---

# Q: What does:

```bash
git rm --cached file.py
```

do?

## Answer

It removes the file from Git tracking but keeps the file locally.

Very useful for:

- `.gitignore`
- large files
- model checkpoints

---

# Q: What does `-r` mean in `git rm -r`?

## Answer

`-r` means recursive.

Used for folders.

Example:

```bash
git rm -r my_folder/
```

---

# Q: What does `-f` or `--force` mean?

## Answer

It forces Git to remove files even if modified.

Dangerous because local work may be lost.

---

# Typical Git Workflow

---

# Q: What is a common Git workflow?

## Answer

```bash
git init
git status
git add .
git commit -m "Initial commit"
```

Then repeatedly:

```bash
git status
git add .
git commit -m "Meaningful message"
```

# Final Mental Model

```text
Working Directory
        ↓
    git add
        ↓
Staging Area
        ↓
   git commit
        ↓
Repository (Commits)
```

# Important Summary

---

# `git add`

Moves changes into staging.

---

# `git commit`

Creates permanent snapshots.

---

# `git restore`

Restores files or unstages files.

---

# `git reset`

Moves HEAD and modifies history.

---

# `git log`

Shows project history.

---

# `git rm`

Removes tracked files safely.

---

# Important Safety Advice

Be very careful with:

```bash
git reset --hard
git rm -f
```

because they may permanently remove work.
