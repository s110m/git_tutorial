# Git Branching

---

# Introduction

One of the most powerful ideas in Git is branching.

Branches allow developers to:

- work on new features safely
- experiment without damaging stable code
- collaborate with teams
- organize development cleanly

Modern software engineering heavily depends on branching.

Without branches, collaborative development would become extremely difficult.

---

# What is a Branch?

---

# Q: What is a branch in Git?

## Answer

A branch is:

- an independent line of development
- a movable pointer to commits
- a separate workspace inside the same repository

---

# Example

Suppose we have:

```text
A -> B -> C (main)
```

Now create a new branch:

```text
A -> B -> C
          \
           feature
```

Both branches initially point to commit `C`.

---

# Q: Why do we use branches?

## Answer

Branches allow us to:

- develop new features safely
- experiment with ideas
- fix bugs independently
- collaborate with teams
- isolate unstable code

without affecting the stable project.

---

# Example

Suppose:

- `main` = stable project
- `feature_ai` = experimental AI feature

If the experimental branch becomes broken, the `main` branch still remains safe.

---

# Branches Are Extremely Common

Many GitHub repositories have:

- several branches
- experimental branches
- development branches
- release branches

Large projects may have dozens or even hundreds of branches.

---

# Creating Branches

---

# Q: How do we create a branch?

## Answer

```bash
git branch new_branch
```

Example:

```bash
git branch feature_transformer
```

---

# Important Idea

When a new branch is created:

> it starts from the current state of the active branch.

---

# Example

Suppose:

```text
A -> B -> C (main)
```

You are currently on `main`.

Now run:

```bash
git branch feature
```

Result:

```text
A -> B -> C
          \
           feature
```

The new branch starts from commit `C`.

---

# Branches from Branches

---

# Q: Can we create branches from branches?

## Answer

Yes — absolutely.

This is very common.

---

# Example

Suppose:

```text
A -> B -> C (main)
          \
           D -> E (feature1)
```

Now while on `feature1`:

```bash
git branch feature2
```

Result:

```text
A -> B -> C (main)
          \
           D -> E
                \
                 feature2
```

This means branches can continue from other branches.

---

# Viewing Branches

---

# Q: How can we see all local branches?

## Answer

```bash
git branch
```

Example output:

```text
* main
  feature1
  feature2
```

The `*` shows the active branch.

---

# Switching Branches

---

# Q: How do we switch between branches?

## Answer

Modern Git command:

```bash
git switch branch_name
```

Example:

```bash
git switch feature1
```

---

# What happens when we switch?

Git changes:

- visible files
- project state
- current working tree

to match that branch.

Amazing!

---

# Example

Suppose:

## Branch `main`

contains:

```text
main.py
README.md
```

---

## Branch `feature_ai`

contains:

```text
main.py
README.md
train_ai.py
```

When switching:

```bash
git switch feature_ai
```

Git changes the visible files automatically.

Now:

```text
train_ai.py
```

appears.

---

# Branch-Specific Files

---

# Q: Can branches contain different files?

## Answer

Yes.

Each branch may contain:

- shared files
- different versions of files
- completely unique files

---

# Example

## Branch `main`

```text
main.py
README.md
```

---

## Branch `feature_web`

```text
main.py
README.md
website.py
```

---

## Branch `feature_ai`

```text
main.py
README.md
train.py
model.py
```

Each branch can evolve independently.

# Shared Files Between Branches

---

# Q: Can files exist in several branches?

## Answer

Yes.

Most files are usually shared.

However:

- their contents may differ
- their versions may differ

between branches.

---

# Example

Suppose:

```python
# main.py in main branch
print("simple version")
```

but in another branch:

```python
# main.py in AI branch
print("AI version")
```

Same filename, different contents.

# Git Clone and Remote Branches

---

# Q: What happens when we clone a GitHub repository?

## Answer

Suppose:

```bash
git clone <repository_url>
```

Git usually checks out:

- the default branch
- usually `main`

So initially you only see the files of the main branch.

---

# Q: Are other branches downloaded during clone?

## Answer

Usually:

- branch information is downloaded
- commits are downloaded

but only the default branch becomes active locally.

---

# Q: How can we see remote branches?

## Answer

Use:

```bash
git branch -r
```

Example:

```text
origin/main
origin/dev
origin/feature_ai
```

This shows remote branches.

---

# Q: How can we see both local and remote branches?

## Answer

```bash
git branch -a
```

Example:

```text
* main
  feature1
  remotes/origin/main
  remotes/origin/dev
```

---

# Switching to Remote Branches

---

# Q: What happens when we switch to a remote branch?

## Answer

Suppose:

```bash
git switch feature_ai
```

Git may:

- create a local tracking branch
- download necessary objects/files
- update working directory

Then the files from that branch appear.

This feels almost magical when first learning Git.

---

# Old vs New Switching Commands

---

# Q: What was used before `git switch`?

## Answer

Previously:

```bash
git checkout branch_name
```

Example:

```bash
git checkout feature1
```

---

# Q: Why is `git switch` better?

## Answer

Because `checkout` was overloaded.

It was used for:

- switching branches
- restoring files
- detached HEAD operations

which confused many beginners.

Modern Git introduced:

```bash
git switch
git restore
```

to make commands clearer.

---

# Switching Branches with Modified Files

---

# Q: Why does Git sometimes prevent switching branches?

## Answer

Suppose you modified:

```text
main.py
```

and this file also exists differently in another branch.

If switching branches would overwrite your changes, Git prevents it.

Example error:

```text
Please commit your changes or stash them before you switch branches.
```

Git protects your work.

---

# Q: Why does Git sometimes allow switching without commit?

## Answer

If the modified file:

- only belongs to the current branch
- or does not conflict with target branch

then switching may be safe.

So Git allows it.

---

# Example

Suppose:

## Branch `feature_ai`

contains:

```text
ai_model.py
```

but branch `main` does NOT contain this file.

Modifying `ai_model.py` may not block switching because there is no conflict with `main`.

---

# Commiting Properly in Branches

---

# Q: Should each branch have its own meaningful commits?

## Answer

Absolutely.

Each branch should have:

- clean history
- meaningful commit messages
- organized development

---

# Example

Bad commit:

```text
update
```

Better commit:

```text
Add transformer attention layer
```

---

# Why is this important?

Good branch history helps:

- collaboration
- debugging
- code review
- merging
- project maintenance

# Typical Branch Workflow

---

# Example Workflow

Create branch:

```bash
git branch feature_ai
```

Switch branch:

```bash
git switch feature_ai
```

Modify files.

Stage changes:

```bash
git add .
```

Commit:

```bash
git commit -m "Add AI training pipeline"
```

Switch back:

```bash
git switch main
```

Git allows moving between these timelines safely.

# Useful Branch Commands

| Command | Purpose |
|---|---|
| `git branch` | show local branches |
| `git branch -r` | show remote branches |
| `git branch -a` | show all branches |
| `git branch new_name` | create branch |
| `git switch branch_name` | switch branch |
| `git switch -c new_name` | create and switch branch |
| `git checkout branch_name` | old switching command |

---

# Important Safety Advice

Before switching branches:

- commit important changes
- or stash them

especially if modified files exist in multiple branches.

Git prevents dangerous overwrites for your safety.
