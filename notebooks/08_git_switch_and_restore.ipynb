# Git `switch` and `restore`

Modern Git introduced:

```bash
git switch
```

and

```bash
git restore
```

to replace many confusing usages of:

```bash
git checkout
```

Because `checkout` was doing TOO MANY different jobs.

---

# Main Idea

## `git switch`
Used for:

- Switching branches
- Moving between branches
- Creating new branches
- Detached HEAD navigation

---

## `git restore`
Used for:

- Restoring files
- Undoing file modifications
- Recovering staged or unstaged files

---

# Why Did Git Create Them?

Old Git:

```bash
git checkout
```

did all of these:

- switch branch
- restore file
- visit old commit
- detached HEAD

This confused many users.

So Git split responsibilities:

| Old | New |
|---|---|
| `checkout branch` | `switch branch` |
| `checkout -- file` | `restore file` |

---

# PART 1 — `git switch`

---

# 1. Switch Branch

Suppose:

```text
A --- B --- C   (main)
       \
         D --- E   (feature)
```

Currently:

```text
HEAD -> main
```

Command:

```bash
git switch feature
```

Result:

```text
HEAD -> feature
```

Now your files become exactly like `feature`.

---

# Analogy

Imagine branches are different rooms.

`git switch` means:

> "Walk into another room."

---

# 2. Create and Switch to New Branch

Old way:

```bash
git checkout -b new_branch
```

Modern way:

```bash
git switch -c new_branch
```

Example:

```bash
git switch -c experiment
```

Result:

```text
A --- B --- C   (main)
             \
               D   (experiment)
```

---

# 3. Detached HEAD with Switch

You can also visit old commits:

```bash
git switch --detach B
```

Result:

```text
A --- B --- C --- D   (main)
      ^
     HEAD
```

You are temporarily exploring old commit `B`.

---

# Important Note

`switch` is mostly for:

```text
branch movement
```

not file restoration.

---

# PART 2 — `git restore`

---

# Main Purpose

`git restore` is for:

```text
Undoing file changes
```

without moving branches.

---

# 1. Restore Modified File

Suppose:

```text
file.py
```

was modified accidentally.

Command:

```bash
git restore file.py
```

Effect:

```text
file.py becomes exactly like last commit
```

Your local modifications disappear.

---

# Analogy

Imagine:

- Git commit = saved game
- Working directory = current gameplay

`restore` means:

> "Reload saved game for this file."

---

# Example

Before:

```text
Last commit:
print("hello")
```

You modify:

```python
print("broken code")
```

Then:

```bash
git restore file.py
```

Result:

```python
print("hello")
```

---

# 2. Restore All Files

```bash
git restore .
```

Effect:

```text
Undo ALL unstaged modifications
```

Dangerous if you have important changes.

---

# 3. Restore Staged Files

Suppose:

```bash
git add file.py
```

Now file is staged.

To unstage:

```bash
git restore --staged file.py
```

Effect:

```text
Removes file from staging area
Keeps file modifications
```

---

# Important Difference

## `restore`

usually affects:

```text
working directory
or
staging area
```

NOT branch history.

---

# Visual Model

Git has 3 important areas:

```text
Working Directory
        ↓
Staging Area
        ↓
Repository (commits)
```

---

# `git restore` Can Operate On:

| Area | Command |
|---|---|
| Working directory | `git restore file.py` |
| Staging area | `git restore --staged file.py` |

---

# Restore From Old Commit

You can restore a file from a specific commit.

Example:

```bash
git restore --source=B file.py
```

Effect:

```text
Copies file.py from commit B
into current working directory
```

---

# Example Diagram

Suppose:

```text
A --- B --- C   (main)
```

Current file:

```python
print("new code")
```

Restore from `B`:

```bash
git restore --source=B file.py
```

Result:

```python
print("old code from B")
```

---

# `switch` vs `restore`

| Feature | `switch` | `restore` |
|---|---|---|
| Changes branch? | Yes | No |
| Restores files? | No | Yes |
| Used for navigation? | Yes | No |
| Used for undoing file edits? | No | Yes |
| Detached HEAD support? | Yes | No |

---

# Relationship with `checkout`

Old Git:

```bash
git checkout feature
git checkout -- file.py
```

Modern Git:

```bash
git switch feature
git restore file.py
```

Cleaner and less confusing.

# Applications of `git switch`

---

## 1. Move Between Features

```bash
git switch feature-login
```

---

## 2. Create Experimental Branches

```bash
git switch -c test-model
```

---

## 3. Explore Old Commits

```bash
git switch --detach a1b2c3
```

## 4. Return to Main Branch

```bash
git switch main
```

# Applications of `git restore`

---

## 1. Undo Bad File Changes

```bash
git restore app.py
```

---

## 2. Unstage Accidentally Added Files

```bash
git restore --staged app.py
```

---

## 3. Recover Old Version of File

```bash
git restore --source=B app.py
```

---

## 4. Undo ALL Local Modifications

```bash
git restore .
```

# Important Warnings

---

# Warning 1

```bash
git restore .
```

can permanently remove local modifications.

---

# Warning 2

`restore` does NOT recover deleted commits.

That is job of:

```bash
git reflog
```

# Q & A

---

## Q1: Is `switch` same as `checkout`?

Partially yes.

```bash
git switch
```

replaces:

```bash
git checkout <branch>
```

---

## Q2: Is `restore` same as `reset`?

Not exactly.

`restore` mainly affects:

- files
- staging area

while `reset` can move branch history.

---

## Q3: Does `restore` delete commits?

No.

It only changes files/staging.

---

## Q4: Can `switch` create branches?

Yes:

```bash
git switch -c new_branch
```

---

## Q5: What does `--staged` mean?

It affects the staging area instead of working directory.

Example:

```bash
git restore --staged file.py
```

---

## Q6: Does `restore` recover deleted branch history?

No.

Use:

```bash
git reflog
```

for that.

---

## Q7: Why did Git split checkout into two commands?

Because `checkout` was overloaded and confusing.

Git wanted:

```text
switch -> branch movement
restore -> file recovery
```
