## Get started: 1- Download and install Git for your OS (Windows, mac, linux). 

2- git --version    This shows that you have installed git properly. 

3- use "git init" in your working folder. This is the start of the journey. 

4- by default, we start a "master" branch. Just one branch at this moment and then later we can start creating new branches as well. 

5- Our goal is to store our files in Github. So, you can create a repository in github and then push your exsitance code into that repository. 

6- or you can "clone" an existance repository by using "git clone <the url of that repo>". Just this. 

7- Note: When you clone a public repository that has several branches, by default you have the files of the "main" branch. But, if you use "git branch -r" then you can see all the branches and by using "git switch <the branch name>, 
then it start to download the files of other repositories as well. 

8- Imagine that you clone a public repository in your own computer. Can you change it and push into it? why not? What is the push request? 



# git status

This command give as the status of uncommited files in the current branch. Usually you are in the "main" branch but it is better that first you check your branch by "git branch" and change your branch if necessary by "git switch <branch name>

# git add --all or git add -A

This put the changes into Stage and ready for commit. 

# git add .
This do for the current directory not all the files. But, if you are in the root directory, it is the same as git add --all

# git add <file name>

We can add some file separately instead of adding all files. 

What is the philosophy of adding into "Stage" and them commit?

1- I guess maybe you have some files or folders that you don't want to commit them at all. For example, you create a file that is for your guidance but you don't want to put it in the project. 

2- By control on adding files separately, you will have a better control on commit and messages that write for it:

git add model.py
git commit -m "Fix bug in model"

git add README.md
git commit -m "Update README formatting"

3- ... 



What does the following do?

git restore --staged file.py
git restore file.py

Answer: The first line, put the file back to the work directory (from Staging directory). then the next line will put it back to the previous commit!! It means if you edit this, then it delete the changes. So, you should be careful about using these two with each other.
