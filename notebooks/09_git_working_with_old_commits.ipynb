# Working with Old Commits in Git

This guide explains common practical questions such as:

- How to view a project at an old commit
- How to compare commits with current code
- How to restore old versions of files
- How to restore only some files from old commits

---

# 1. How to See the Project at a Specific Commit Without Changing Anything

Suppose your history is:

```text
A --- B --- C --- D   (main)
```

You are currently at:

```text
D
```

and want to temporarily inspect commit `B`.

---

# Method 1 — Detached HEAD (`switch --detach`)

```bash
git switch --detach B
```

Result:

```text
A --- B --- C --- D   (main)
      ^
     HEAD
```

What happens:

- Your files become exactly like commit `B`
- `main` branch DOES NOT move
- History is unchanged
- You are temporarily visiting old code

---

# Analogy

Imagine:

- The branch is a bookmark in a book
- You temporarily go back and read old pages
- The bookmark itself does NOT move

---

# Return Back

```bash
git switch main
```

---

In VSCode:

```text
GitLens extension
```

lets you inspect old commits visually.

# 3. How to Return a File to an Old Commit Version

Suppose:

```text
Current app.py is broken
```

and you want version from commit `B`.

---

# Method — `git restore`

```bash
git restore --source=B app.py
```

Result:

```text
Current app.py becomes version from commit B
```

Important:

- Branch history does NOT change
- Only file content changes

---

# Diagram

Before:

```text
Current file:
new broken code
```

Restore:

```bash
git restore --source=B app.py
```

After:

```text
old stable code from B
```

---

# Commit the Restoration

After restoring:

```bash
git add app.py
git commit -m "Restore app.py from commit B"
```

---

# 4. How to Restore Only SOME Files from Old Commit

This is extremely common in real projects.

Suppose:

```text
Commit B has:
- good app.py
- bad model.py
```

You only want:

```text
app.py
```

---

# Restore Specific Files

```bash
git restore --source=B app.py
```

or multiple files:

```bash
git restore --source=B app.py utils.py train.py
```

---

# What Happens?

Only selected files are restored.

Everything else stays unchanged.

---

# Analogy

Imagine old backup archive.

You do NOT restore the whole computer.

You only restore:

```text
selected files
```

from backup.

---

# 5. How to Restore Entire Project from Old Commit

If you truly want everything from old commit:

---

# Option 1 — Temporary Exploration

```bash
git switch --detach B
```

Safe.

History unchanged.

---

# Option 2 — Permanent Reset

```bash
git reset --hard B
```

Dangerous.

Result:

```text
A --- B   (main)
```

Commits after `B` disappear from branch history.

---

# Important Difference

| Command | Safe? | Changes History? |
|---|---|---|
| `switch --detach` | Yes | No |
| `restore` | Yes | No |
| `reset --hard` | Dangerous | Yes |

# Q & A

---

## Q1: Does `switch --detach` modify branch history?

No.

It is temporary exploration only.

---

## Q2: Does `restore` move branches?

No.

It only changes file contents.

---

## Q3: Can I restore only one file from old commit?

Yes.

```bash
git restore --source=B app.py
```

---

## Q4: Can I compare old commit with current project?

Yes.

```bash
git diff B
```

or use VSCode GitLens.

---

## Q5: What if I accidentally restore wrong file?

Before commit:

```bash
git restore app.py
```

can usually undo unstaged changes.

---

## Q6: What if I want entire project from old commit permanently?

Use:

```bash
git reset --hard B
```

carefully.

---

## Q7: What is safest method for exploration?

```bash
git switch --detach
```

because it changes nothing permanently.
