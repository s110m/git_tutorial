# GitHub Pull Request: Step-by-Step Guide

A **Pull Request** is a way to say:

> "I changed something in a separate branch. Please review it and merge it into the main project."

It is one of the most important workflows in GitHub team projects.

---

# 1. Why Do We Need Pull Requests?

Without Pull Requests, people may push directly to:

```text
main
```

This can be dangerous because:

- broken code may enter the main project
- nobody reviews the changes
- tests may not run before merging
- team members may overwrite each other's work
- project history becomes messy

---

# Pull Request Idea

Instead of changing `main` directly:

```text
main branch = stable official project
feature branch = your working area
```

You work on your own branch, then ask:

> "Can we merge this branch into main?"

That question is the Pull Request.

---

# Analogy

Imagine a company document.

`main` is the official final document.

You do NOT directly edit the final version.

Instead:

1. copy the document
2. make your changes
3. submit your version for review
4. manager checks it
5. if good, manager adds it to the official document

That review request is like a Pull Request.

---

# 2. First Step: Clone the Repository or Not?

It depends.

---

# Case 1 — You Are Starting From GitHub

If the project already exists on GitHub, then YES, usually the first step is:

```bash
git clone https://github.com/username/project.git
```

This downloads the repository to your computer.

---

# Case 2 — You Already Have the Project Locally

If you already cloned it before, you do NOT clone again.

You just go inside the project folder:

```bash
cd project
```

Then update it:

```bash
git pull
```

or safer:

```bash
git fetch
git pull
```

---

# Case 3 — You Do Not Have Permission to Push

For public open-source repositories, you may not have write access.

Then you usually:

```text
Fork the repository first
```

Then clone your fork.

---

# 3. Basic Pull Request Workflow

The usual workflow is:

```text
1. Clone repository
2. Create new branch
3. Make changes
4. Commit changes
5. Push branch to GitHub
6. Open Pull Request
7. Review and discuss
8. Merge Pull Request
```

---

# 4. Step-by-Step Commands

---

# Step 1 — Clone Repository

```bash
git clone https://github.com/company/project.git
cd project
```

Now you have local copy.

Diagram:

```text
GitHub Repository
        |
        | clone
        v
Your Laptop
```

---

# Step 2 — Make Sure You Are Updated

```bash
git switch main
git pull
```

This makes sure your local `main` is latest.

---

# Step 3 — Create a New Branch

```bash
git switch -c feature-login
```

Meaning:

```text
Create branch feature-login and move into it
```

Diagram:

```text
A --- B --- C   main
             \
              feature-login
```

---

# Step 4 — Make Changes

Edit files.

Example:

```text
login.py
README.md
tests/test_login.py
```

Then check status:

```bash
git status
```

---

# Step 5 — Stage and Commit

```bash
git add .
git commit -m "Add login feature"
```

Now your branch has a new commit.

Diagram:

```text
A --- B --- C   main
             \
              D   feature-login
```

---

# Step 6 — Push Your Branch to GitHub

```bash
git push -u origin feature-login
```

Meaning:

```text
Upload my feature-login branch to GitHub
```

Now GitHub has:

```text
origin/main
origin/feature-login
```

---

# Step 7 — Create Pull Request on GitHub

Go to GitHub.

Usually GitHub shows a button:

```text
Compare & pull request
```

Click it.

Choose:

```text
base branch: main
compare branch: feature-login
```

Meaning:

```text
I want to merge feature-login into main
```

---

# Pull Request Direction

Very important:

```text
feature-login  --->  main
```

Not the opposite.

Diagram:

```text
feature-login changes
        |
        | Pull Request
        v
main
```

---

# Step 8 — Write Pull Request Description

Example:

```text
Title:
Add login feature

Description:
This PR adds login functionality, input validation, and basic tests.
```

A good Pull Request explains:

- what changed
- why it changed
- how to test it
- any risks or limitations

---

# 5. What Happens After Creating Pull Request?

After PR is created:

```text
GitHub compares branches
```

It shows:

- changed files
- added lines
- deleted lines
- commits
- comments
- conflicts if any
- automated test results if configured

---

# Pull Request Page Shows

```text
Conversation
Commits
Checks
Files changed
```

---

# 6. Review Process

Other developers or project owners can:

- read your code
- leave comments
- request changes
- approve the PR
- run tests
- discuss implementation

---

# Analogy

Pull Request is like saying:

> "Here is my draft. Please check before we add it to the official version."

---

# 7. What If Reviewer Requests Changes?

You do NOT create a new Pull Request usually.

You just continue working on the SAME branch.

Example:

```bash
git switch feature-login
```

Make changes.

Then:

```bash
git add .
git commit -m "Fix review comments"
git push
```

GitHub automatically updates the Pull Request.

---

# Important

A Pull Request follows the branch.

So every new commit pushed to that branch appears in the same PR.

---

# Diagram

Before review fix:

```text
A --- B --- C   main
             \
              D   feature-login
```

After fix:

```text
A --- B --- C   main
             \
              D --- E   feature-login
```

The PR updates automatically.

---

# 8. What If There Are Conflicts?

Suppose `main` changed after your branch was created.

```text
A --- B --- C --- F   main
             \
              D --- E   feature-login
```

Now GitHub may say:

```text
This branch has conflicts that must be resolved
```

---

# How to Handle Conflict Locally

Update your branch:

```bash
git fetch
git switch feature-login
git merge origin/main
```

Resolve conflicts manually.

Then:

```bash
git add .
git commit
git push
```

---

# Alternative With Rebase

```bash
git fetch
git switch feature-login
git rebase origin/main
```

If conflict happens:

```bash
git add .
git rebase --continue
```

Then push:

```bash
git push --force-with-lease
```

Use rebase carefully, especially if others also work on your branch.

---

# 9. If You Are the Repository Owner, What Should You Do?

If you are owner or maintainer, you review the Pull Request.

---

# Owner Checklist

Check:

```text
1. Does the code solve the problem?
2. Is the code understandable?
3. Are there tests?
4. Do tests pass?
5. Are there conflicts?
6. Is the branch safe to merge?
7. Is the PR description clear?
```

---

# Owner Options on GitHub

You can:

```text
Comment
Approve
Request changes
Close PR
Merge PR
```

---

# 10. Merge Pull Request

When everything is good, click:

```text
Merge pull request
```

GitHub usually gives options:

---

# Option 1 — Create a Merge Commit

```text
Preserves branch history
```

Diagram:

```text
A --- B --- C -------- M   main
             \        /
              D --- E
```

Good when you want to preserve full branch history.

---

# Option 2 — Squash and Merge

```text
Combines all PR commits into one commit
```

Diagram:

```text
A --- B --- C --- S   main
```

`S` = one squashed commit.

Good when branch has many messy commits like:

```text
fix
fix again
small change
test
```

---

# Option 3 — Rebase and Merge

```text
Replays commits onto main linearly
```

Diagram:

```text
A --- B --- C --- D --- E   main
```

Good for clean linear history.

---

# 11. After PR Is Merged

Usually delete the feature branch on GitHub:

```text
Delete branch
```

Then update your local machine:

```bash
git switch main
git pull
```

Delete local branch:

```bash
git branch -d feature-login
```

---

# 12. If You Are Contributor Without Write Access

For open-source projects, workflow is slightly different.

---

# Fork Workflow

```text
Original repo
     |
     | fork
     v
Your GitHub copy
     |
     | clone
     v
Your laptop
```

Commands:

```bash
git clone https://github.com/yourname/project.git
cd project
git switch -c fix-bug
```

Then:

```bash
git add .
git commit -m "Fix bug"
git push -u origin fix-bug
```

Then open PR from:

```text
your fork/fix-bug
```

to:

```text
original repo/main
```

---

# 13. Pull Request vs Push

| Action | Meaning |
|---|---|
| `git push` | Upload your branch to GitHub |
| Pull Request | Ask to merge your branch into another branch |

Important:

```text
push does not automatically merge into main
```

It only uploads your branch.

---

# 14. Pull Request vs Pull

Do not confuse:

```bash
git pull
```

with:

```text
Pull Request
```

They are different.

---

# `git pull`

Means:

```text
download and combine remote changes into local branch
```

---

# Pull Request

Means:

```text
request to merge one branch into another on GitHub
```

---

# 15. Common Professional Workflow

```bash
git clone https://github.com/company/project.git
cd project

git switch main
git pull

git switch -c feature-login

# edit files

git add .
git commit -m "Add login feature"

git push -u origin feature-login
```

Then on GitHub:

```text
Open Pull Request:
feature-login -> main
```

# 16. Q & A

---

## Q1: Is the first step always clone?

No.

If you already have the repository locally, do not clone again.

Just update:

```bash
git pull
```

---

## Q2: Do I create PR from main?

Usually no.

You create PR from your feature branch into main.

---

## Q3: Does push create Pull Request automatically?

No.

Push only uploads branch.

You create PR on GitHub.

---

## Q4: Can I update PR after creating it?

Yes.

Push more commits to the same branch.

---

## Q5: Who merges Pull Request?

Usually repository owner, maintainer, or someone with permission.

---

## Q6: What if I am the owner?

You review the code, check tests/conflicts, then merge if good.

---

## Q7: What if there are conflicts?

Resolve conflicts locally, then push again.

---

## Q8: Should I delete branch after merge?

Usually yes.

The work is already in main.

---

## Q9: What is base branch?

The branch you want to merge INTO.

Usually:

```text
main
```

---

## Q10: What is compare branch?

The branch containing your changes.

Example:

```text
feature-login
```

---

## Q11: Can one PR contain many commits?

Yes.

All commits from the compare branch appear in the PR.

---

## Q12: Can one PR change many files?

Yes.

But smaller PRs are easier to review.
