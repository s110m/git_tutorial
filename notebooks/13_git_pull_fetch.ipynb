# Understanding `git pull`, `git fetch`, and `--rebase`

One of the most important concepts in Git team workflow is understanding:

```bash
git fetch
git pull
git pull --rebase
```

Many Git problems and conflicts become much easier once you understand the difference between these commands.

---

# Big Picture

Suppose:

```text
Your Laptop  <------>  GitHub Repository
```

Other developers may push changes to GitHub.

Your local repository may become:

```text
outdated
```

You need ways to:

- download updates
- inspect updates
- merge updates
- rebase updates

---

# Main Difference

| Command | What It Does |
|---|---|
| `git fetch` | Download remote updates ONLY |
| `git pull` | Download + merge |
| `git pull --rebase` | Download + rebase |

---

# 1. Understanding `git fetch`

---

# Main Idea

```bash
git fetch
```

means:

> "Download latest information from remote repository, but DO NOT modify my working branch yet."

---

# Important

`fetch` is SAFE.

It does NOT:

- modify your files
- merge changes
- create conflicts immediately

---

# Visual Example

Suppose GitHub has:

```text
A --- B --- C   (origin/main)
```

Your local repository:

```text
A --- B   (main)
```

You run:

```bash
git fetch
```

Result:

```text
Local main:
A --- B

Remote tracking:
A --- B --- C   (origin/main)
```

---

# Important Observation

After fetch:

```text
main != origin/main
```

Your local branch is still unchanged.

---

# Analogy

Imagine:

```text
GitHub = central company archive
```

`fetch` means:

> "Download latest archive information, but don't touch my current work yet."

---

# What Does Fetch Update?

Fetch updates:

```text
remote-tracking branches
```

such as:

```text
origin/main
origin/feature
```

---

# How to See Difference After Fetch

Example:

```bash
git diff main origin/main
```

or:

```bash
git log main..origin/main
```

---

# When Do We Use Fetch?

---

# Good Use Cases

---

## 1. Safely Inspect Remote Changes

```bash
git fetch
```

before deciding what to do.

---

## 2. Before Important Merge

You want to see incoming commits first.

---

## 3. Team Development

Very common professional workflow.

---

# 2. Understanding `git pull`

---

# Main Idea

```bash
git pull
```

means:

```text
git fetch
+
git merge
```

Git automatically:

1. downloads changes
2. merges them into current branch

---

# Visual Example

Before:

Local:

```text
A --- B --- C   (main)
```

GitHub:

```text
A --- B --- D   (origin/main)
```

You run:

```bash
git pull
```

Git does:

```text
fetch + merge
```

Result:

```text
A --- B --- D --- M
             \   /
               C
```

`M` = merge commit.

---

# Important Idea

`pull` modifies your current branch.

Unlike `fetch`.

---

# Analogy

`fetch`:

> "Download new blueprints."

`pull`:

> "Download AND immediately combine with my work."

---

# Possible Errors During Pull

YES.

Very common.

---

# Scenario 1 — Merge Conflict

Suppose:

You changed:

```python
line 10
```

Other developer ALSO changed:

```python
same line 10
```

Git cannot decide.

---

# Conflict Example

Git creates:

```python
<<<<<<< HEAD
print("Ali version")
=======
print("Sara version")
>>>>>>> main
```

---

# What Should You Do?

Manually edit file.

Then:

```bash
git add .
git commit
```

---

# Scenario 2 — Uncommitted Local Changes

Suppose:

You modified files locally but did NOT commit.

Then:

```bash
git pull
```

may fail.

---

# Error Example

```text
Your local changes would be overwritten by merge.
```

---

# Why?

Git protects your uncommitted work.

---

# Solutions

---

## Option 1 — Commit Changes

```bash
git add .
git commit -m "WIP"
git pull
```

---

## Option 2 — Stash Changes

```bash
git stash
git pull
git stash pop
```

---

# What is `git stash`?

Temporary hidden storage for uncommitted changes.

Analogy:

> "Put unfinished papers in drawer temporarily."

---

# Scenario 3 — Divergent Branches

Suppose:

GitHub:

```text
A --- B --- D
```

Local:

```text
A --- B --- C
```

Git may ask:

```text
Need to specify how to reconcile divergent branches.
```

Modern Git sometimes requires explicit choice:

- merge
- rebase
- fast-forward only

---

# 3. Understanding `git pull --rebase`

---

# Main Idea

Instead of merging:

```text
rebase rewrites your commits
on top of latest remote history
```

---

# Normal Pull (Merge)

Before:

```text
Local:
A --- B --- C

Remote:
A --- B --- D
```

After pull:

```text
A --- B --- D --- M
             \   /
               C
```

Creates merge commit `M`.

---

# Pull With Rebase

Command:

```bash
git pull --rebase
```

Result:

```text
A --- B --- D --- C'
```

Your commit is replayed on top of newest remote commit.

---

# Important

`C'` means:

```text
rebuilt version of C
```

Technically new commit.

---

# Analogy

Suppose:

You wrote report page `C`.

Then another developer inserted page `D`.

Instead of creating merge page:

Git says:

> "Rewrite your page after D."

Cleaner history.

---

# Why Many Developers Prefer Rebase

Rebase gives:

```text
cleaner linear history
```

instead of many merge commits.

---

# Merge History

```text
A --- B --- D --- M
             \   /
               C
```

Messier.

---

# Rebase History

```text
A --- B --- D --- C'
```

Cleaner.

---

# What Happens Internally During Rebase?

Git temporarily:

1. removes your commits
2. updates branch to latest remote
3. reapplies your commits one-by-one

---

# Can Rebase Also Create Conflicts?

YES.

Very common.

---

# Example

Suppose both developers modified same lines.

During replay:

```text
Git cannot apply your commit cleanly.
```

Conflict appears.

---

# What Should You Do?

---

# Step 1 — Resolve Conflict

Edit files manually.

---

# Step 2 — Add Files

```bash
git add .
```

---

# Step 3 — Continue Rebase

```bash
git rebase --continue
```

---

# Abort Rebase

If everything becomes messy:

```bash
git rebase --abort
```

---

# Important Warning About Rebase

Avoid rebasing:

```text
shared public commits
```

because rebase rewrites commit history.

---

# Good Rule

Safe:

```text
rebase your own local commits
```

Dangerous:

```text
rebase already shared commits
```

---

# Fetch vs Pull vs Pull --Rebase

| Command | Downloads | Modifies Current Branch | Creates Merge Commit |
|---|---|---|---|
| `fetch` | Yes | No | No |
| `pull` | Yes | Yes | Usually yes |
| `pull --rebase` | Yes | Yes | Usually no |

---

# Common Professional Workflow

Very common modern workflow:

```bash
git fetch
git rebase origin/main
```

or:

```bash
git pull --rebase
```

This keeps history cleaner.

---

# Example Real Workflow

---

# Step 1

Get latest changes:

```bash
git fetch
```

---

# Step 2

Inspect changes:

```bash
git log HEAD..origin/main
```

---

# Step 3

Rebase:

```bash
git rebase origin/main
```

---

# Step 4

Resolve conflicts if needed:

```bash
git add .
git rebase --continue
```

---

# Analogy of Engineers

Imagine engineers editing blueprints.

---

# Fetch

Engineer says:

> "Download latest company blueprint updates."

No local changes yet.

---

# Pull

Engineer says:

> "Download and immediately merge updates into my work."

---

# Rebase

Engineer says:

> "Take my unfinished edits and rewrite them on top of latest official blueprint."

# Q & A

---

## Q1: Does `git fetch` change my files?

No.

It only updates remote-tracking information.

---

## Q2: Does `git pull` modify my branch?

Yes.

It performs fetch + merge.

---

## Q3: Is `fetch` safer than `pull`?

Yes.

Because it changes nothing locally.

---

## Q4: Why do merge conflicts happen?

Usually because:

```text
same lines were modified differently
```

---

## Q5: What is advantage of rebase?

Cleaner linear history.

---

## Q6: Does rebase create new commits?

Yes.

Rebased commits are technically rebuilt commits.

---

## Q7: What should I do after conflict during rebase?

```bash
git add .
git rebase --continue
```

---

## Q8: How do I cancel rebase?

```bash
git rebase --abort
```

---

## Q9: Should I always use rebase?

Not always.

Depends on team workflow.

Many teams prefer:

```bash
git pull --rebase
```

for cleaner history.

---

## Q10: Is merge bad?

No.

Merge preserves true historical branching.
