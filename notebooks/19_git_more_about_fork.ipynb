# Understanding Forks, Branches, Pushes, and Pull Requests on GitHub

## Motivation

Many beginners are confused about the relationship between:

- Fork
- Clone
- Branch
- Push
- Pull Request (PR)

A common misunderstanding is:

> "When I push a branch, GitHub automatically creates a Pull Request."

or

> "If I create a Pull Request from my fork, does it immediately merge into the original repository?"

This note explains the complete workflow.

---

# 1. What is a Fork?

Suppose there is a public repository:

```text
owner/project
```

You do not own it.

A **fork** is a copy of that repository under your GitHub account.

After clicking **Fork**, GitHub creates:

```text
owner/project      (original repository)

seyedali/project   (your fork)
```

You now own:

```text
seyedali/project
```

and can freely:

- create branches
- push commits
- open pull requests

without affecting the original repository.

---

# 2. What is a Clone?

A fork exists on GitHub.

To work on it locally, clone it:

```bash
git clone https://github.com/seyedali/project.git
```

Now you have:

```text
Local Computer
       |
       v
seyedali/project (GitHub fork)
```

---

# 3. Should I Work on main?

Technically:

```text
main
```

can be modified directly.

However, the standard practice is:

```bash
git checkout -b my_feature
```

which creates:

```text
main
my_feature
```

Advantages:

- Keeps main clean
- Easier to create PRs
- Easier to abandon experiments
- Matches professional workflows

---

# 4. Commit Changes

After modifying files:

```bash
git add .
git commit -m "Add new feature"
```

Your changes are now stored locally.

---

# 5. Push Changes to Your Fork

Push the branch:

```bash
git push -u origin my_feature
```

The option:

```bash
-u
```

means:

```text
--set-upstream
```

and records the connection:

```text
local my_feature
       |
       v
origin/my_feature
```

After the push:

```text
Local:

my_feature

        |
        v

GitHub Fork:

main
my_feature
```

---

# 6. Does git push Create a Pull Request?

No.

This is one of the most common misconceptions.

The command:

```bash
git push -u origin my_feature
```

only creates or updates:

```text
seyedali/project:my_feature
```

It does NOT create a Pull Request.

---

# 7. Why Does GitHub Show "Compare & Pull Request"?

After pushing, GitHub often displays:

```text
Compare & pull request
```

This is only a suggestion.

GitHub is saying:

> "I see a new branch. Would you like to create a Pull Request?"

At this point:

```text
PR count = 0
```

No Pull Request exists yet.

---

# 8. What is a Pull Request?

A Pull Request is a request to merge:

```text
Source Branch
      |
      v
Target Branch
```

Example:

```text
my_feature
    |
    v
main
```

A PR is a GitHub feature.

It is NOT a Git command.

---

# 9. Two Types of Pull Requests

## Type A: Inside Your Fork

```text
seyedali/project:my_feature
            |
            v
seyedali/project:main
```

Meaning:

> Merge my_feature into main inside my own fork.

Only your fork changes.

The original repository is unaffected.

---

## Type B: Contribution to Original Repository

```text
seyedali/project:my_feature
            |
            v
owner/project:main
```

Meaning:

> Please merge my changes into the original project.

This is the standard open-source workflow.

---

# 10. Where Do I Create the Pull Request?

Usually from your fork:

```text
github.com/seyedali/project
```

GitHub shows:

```text
Compare & pull request
```

When clicked, GitHub usually suggests:

```text
Base Repository:
owner/project

Base Branch:
main

Head Repository:
seyedali/project

Compare Branch:
my_feature
```

Then clicking:

```text
Create Pull Request
```

creates:

```text
seyedali/project:my_feature
            |
            v
owner/project:main
```

---

# 11. Do I Need Permission to Create a Pull Request?

Usually:

```text
YES
```

for public repositories.

You typically do NOT need write access to create a PR.

You only need:

- your fork
- your branch

The maintainers later decide whether to:

- merge
- reject
- comment
- request changes

---

# 12. Does the PR Immediately Appear in the Original Repository?

Yes.

As soon as you click:

```text
Create Pull Request
```

the PR appears in:

```text
owner/project
```

under:

```text
Pull Requests
```

The maintainers do NOT need to approve its existence.

They only decide whether to merge it.

---

# 13. Can I Close My Pull Request?

Yes.

If you created:

```text
seyedali/project:my_feature
            |
            v
owner/project:main
```

you can close it before it is merged.

Status changes:

```text
Open
```

to

```text
Closed
```

---

# 14. If I Close It, Does It Disappear?

No.

GitHub keeps the record.

The PR remains visible as:

```text
Closed
```

The maintainers can still see:

- comments
- commits
- discussions
- changed files

---

# 15. Does Closing a PR Affect the Original Repository?

No.

Closing a PR means:

```text
"Please ignore my proposal."
```

It does NOT modify:

```text
owner/project:main
```

---

# 16. What Happens When a PR Is Merged?

Before:

```text
owner/project:main

A --- B --- C
```

PR:

```text
seyedali/project:my_feature

A --- B --- C --- D --- E
```

After merge:

```text
owner/project:main

A --- B --- C --- D --- E
```

Now the original repository contains your changes.

---

# Recommended Workflow

```bash
# Fork on GitHub

# Clone your fork
git clone https://github.com/seyedali/project.git

# Create a branch
git checkout -b my_feature

# Make changes

git add .
git commit -m "Add feature"

# Push to your fork
git push -u origin my_feature

# Go to GitHub
# Click "Compare & pull request"

# Create PR:
# my_feature -> owner/main
```

---

# Frequently Asked Questions (Q&A)

## Q1: Does `git push` create a Pull Request?

No.

It only pushes commits and branches.

---

## Q2: Why does GitHub show "Compare & pull request"?

It is only a suggestion.

No PR exists until you click:

```text
Create Pull Request
```

---

## Q3: Should I work directly on `main`?

You can.

However, best practice is:

```bash
git checkout -b my_feature
```

and work on a separate branch.

---

## Q4: Do I need permission to push to the original repository?

Usually yes.

Without write access:

```bash
git push origin my_feature
```

to the original repository will fail.

---

## Q5: Do I need permission to create a Pull Request?

Usually no.

For public repositories, anyone can usually create a PR.

---

## Q6: Where is the Pull Request created?

Usually from your fork's GitHub page.

The PR then targets:

```text
owner/project
```

---

## Q7: Can I create a PR inside my own fork?

Yes.

Example:

```text
my_feature
    |
    v
main
```

inside:

```text
seyedali/project
```

---

## Q8: If I merge a PR inside my fork, does it affect the original repository?

No.

Only your fork changes.

---

## Q9: If I close my PR, does it disappear?

No.

It remains visible as:

```text
Closed
```

---

## Q10: Can I delete my branch after creating a PR?

Yes.

However, reopening the PR later may become difficult if the branch is removed.

---

## Q11: Who can merge the PR?

For:

```text
seyedali/project:my_feature
            |
            v
owner/project:main
```

usually the maintainers of:

```text
owner/project
```

decide whether to merge it.

---

## Q12: What is the most common workflow in open source?

```text
Fork
  ↓
Clone
  ↓
Create Branch
  ↓
Commit
  ↓
Push
  ↓
Create Pull Request
  ↓
Review
  ↓
Merge
```
