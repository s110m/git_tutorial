# Git `restore` vs `reset`

---

# Introduction

Git has several commands for undoing changes, restoring files, or moving backward in history. Two of the most important and commonly confused commands are:

- `git restore`
- `git reset`

Although they may sometimes appear to do similar things, they actually serve very different purposes.

---

# Git Mental Model

Git can be viewed as three areas:

```text
Repository (Commits)  <->  Staging Area  <->  Working Directory
```

## 1. Repository (Commits)

This is the Git history.

Example:

```text
A -> B -> C (HEAD)
```

---

## 2. Staging Area

The place where files are prepared before commit.

Example:

```bash
git add file.py
```

---

## 3. Working Directory

Your actual files on disk.

---

# What is `git restore`?

`git restore` is mainly used for:

- restoring files
- discarding local changes
- unstaging files
- recovering files from old commits

It mainly works with:

```text
Staging Area <-> Working Directory
```

and sometimes copies files from old commits.

---

# What is `git reset`?

`git reset` is mainly used for:

- moving HEAD
- undoing commits
- rewriting local history
- modifying the staging area

It mainly works with:

```text
Repository (Commits) <-> Staging Area
```

and optionally the working directory.

---

# Q&A Section

---

# Q: What is the difference between `restore` and `reset`?

## Answer

### `git restore`

Focuses on:

- files
- working directory
- staging area

It restores file contents.

---

### `git reset`

Focuses on:

- commits
- branch history
- HEAD movement

It changes Git history position.

---

# Q: When do we use `restore` and when `reset`?

## Use `restore` when:

- you accidentally modified a file
- you deleted a file accidentally
- you want to unstage files
- you want to recover a file from an older commit

---

## Use `reset` when:

- you want to undo commits
- you want to move branch history backward
- you want to rewrite local history
- you want to remove commits

---

# Q: What happens with `git restore file.py`?

## Answer

```bash
git restore file.py
```

This means:

> Restore `file.py` from the current commit (`HEAD`).

Example:

Suppose:

```python
print("new change")
```

was added accidentally.

Then:

```bash
git restore file.py
```

removes the local modifications.

---

# Q: What happens with `git restore --staged file.py`?

## Answer

```bash
git restore --staged file.py
```

This removes the file from staging.

Example:

```bash
git add file.py
```

Then:

```bash
git restore --staged file.py
```

Result:

- file changes still exist
- but the file is no longer staged

---

# Q: What is the difference between:

```bash
git restore file.py
```

and

```bash
git restore --staged file.py
```

## Answer

| Command | Effect |
|---|---|
| `git restore file.py` | discard local file modifications |
| `git restore --staged file.py` | remove file from staging |

---

# Q: I committed a deletion by mistake. How can I restore the file?

## Answer

Suppose:

```text
Commit A -> file exists
Commit B -> file deleted
```

Current HEAD is `B`.

You can restore the file from the previous commit:

```bash
git restore --source=HEAD~1 file.py
```

Then commit it again:

```bash
git add file.py
git commit -m "Restore file.py"
```

---

# Q: Can we restore a file from any older commit?

## Answer

Yes.

First find the commit:

```bash
git log -- file.py
```

Then restore from a specific commit:

```bash
git restore --source=<commit-hash> file.py
```

Example:

```bash
git restore --source=a1b2c3d file.py
```

This copies the old version of the file into the working directory.

---

# Q: Does `git restore` undo commits?

## Answer

No.

`git restore` does NOT move Git history backward.

It only restores file contents.

Example:

```bash
git restore .
```

does NOT remove commits.

---

# Q: What does:

```bash
git reset HEAD~1
```

do?

## Answer

This is equivalent to:

```bash
git reset --mixed HEAD~1
```

It:

- removes the last commit
- moves HEAD backward
- keeps file changes

Example:

Before:

```text
A -> B -> C (HEAD)
```

After:

```text
A -> B (HEAD)
```

But the changes from `C` still remain in your files.

---

# Q: What does:

```bash
git reset --hard HEAD~1
```

do?

## Answer

This is much stronger.

It:

- removes the last commit
- moves HEAD backward
- deletes local file changes
- resets working directory completely

Example:

Before:

```text
A -> B -> C (HEAD)
```

After:

```text
A -> B (HEAD)
```

and all changes from `C` disappear from disk.

---

# Q: Why should we be careful with `--hard`?

## Answer

Because it can permanently delete work.

Example:

```bash
git reset --hard HEAD~1
```

may destroy:

- uncommitted changes
- recent commits
- local work

Many beginners accidentally lose files using `--hard`.

---

# Q: What does `HEAD~1` mean?

## Answer

It means:

> One commit before current HEAD.

Example:

```text
A -> B -> C (HEAD)
```

Then:

```text
HEAD~1 = B
```

---

# Q: What are the three reset modes?

## Answer

---

## 1. Soft reset

```bash
git reset --soft HEAD~1
```

Result:

- commit removed
- changes kept
- changes still staged

---

## 2. Mixed reset (default)

```bash
git reset HEAD~1
```

Result:

- commit removed
- changes kept
- changes unstaged

---

## 3. Hard reset

```bash
git reset --hard HEAD~1
```

Result:

- commit removed
- changes deleted

---

# Reset Comparison Table

| Command | Commit Removed | Changes Kept | Staged |
|---|---|---|---|
| `--soft` | Yes | Yes | Yes |
| `--mixed` | Yes | Yes | No |
| `--hard` | Yes | No | No |

---

# Q: I used:

```bash
git reset --hard HEAD~1
```

by mistake. Can I recover?

## Answer

Usually yes.

Use:

```bash
git reflog
```

Example output:

```text
a1b2c3d HEAD@{0}: reset: moving to HEAD~1
f4e5d6g HEAD@{1}: commit: Added tutorial
```

Recover using:

```bash
git reset --hard HEAD@{1}
```

or:

```bash
git reset --hard f4e5d6g
```

---

# Q: What is `git reflog`?

## Answer

`git reflog` shows the history of HEAD movements.

It helps recover:

- lost commits
- bad resets
- accidental history changes

It is one of the most useful Git recovery tools.

---

# Q: What is the difference between:

```bash
git add .
```

and

```bash
git add *
```

## Answer

---

## `git add .`

Recommended.

Adds:

- current directory
- subdirectories
- hidden files

---

## `git add *`

Uses shell wildcard expansion.

Potential issues:

- hidden files may be ignored
- behavior depends on shell
- less reliable

Usually prefer:

```bash
git add .
```

# Important Safety Advice

Be very careful with:

```bash
git reset --hard
```

because it can permanently remove work.

When unsure, prefer safer commands like:

```bash
git restore
git reset --mixed
```

instead of `--hard`.
