# Working with Other Developers in the Same GitHub Repository

Imagine:

- You and several other developers
- are working on the SAME GitHub repository
- everyone has their own local copy
- everyone can push changes

Now suppose:

```bash
git push
```

fails.

Why?

What kinds of errors can happen?

How does Git handle multiple developers changing the same files?

This guide explains the common scenarios.

---

# Big Picture

Each developer has:

```text
Local Repository
```

and everyone connects to:

```text
Shared GitHub Repository
```

Diagram:

```text
Ali's Laptop  ----\
                    \
Sara's Laptop -------> GitHub Repository
                    /
Reza's Laptop -----/
```

---

# Important Idea

GitHub repository is:

```text
shared history
```

Multiple people may:

- modify same files
- create new commits
- push updates

at the same time.

---

# Basic Workflow

Usually developers do:

```bash
git pull
# modify files
git add .
git commit
git push
```

---

# Scenario 1 — No Problem

Suppose history:

```text
A --- B   (GitHub)
```

You pull latest code.

Then create commit:

```text
A --- B --- C   (your local)
```

Nobody else changed GitHub meanwhile.

Then:

```bash
git push
```

works perfectly.

---

# Scenario 2 — Push Rejected

Very common.

Suppose:

---

# Initial State

GitHub:

```text
A --- B
```

You clone repository.

Your local:

```text
A --- B
```

---

# You Create Commit

Local:

```text
A --- B --- C
```

---

# Another Developer Pushes First

GitHub becomes:

```text
A --- B --- D
```

Now histories differ.

---

# You Try Push

```bash
git push
```

ERROR:

```text
! [rejected] main -> main (non-fast-forward)
error: failed to push some refs
```

---

# Why Did This Happen?

Because GitHub says:

> "Your branch is behind remote branch."

Git prevents overwriting others' work.

---

# Visual Diagram

Your local:

```text
A --- B --- C
```

GitHub:

```text
A --- B --- D
```

Git sees:

```text
two different futures
```

---

# What Should You Do?

Usually:

```bash
git pull
```

or better:

```bash
git pull --rebase
```

---

# What Happens During Pull?

Git downloads remote changes and combines histories.

Result:

```text
A --- B --- D --- M
              \   /
                C
```

`M` = merge commit.

---

# Analogy

Imagine two engineers editing same blueprint.

Engineer Ali:

```text
adds engine changes
```

Engineer Sara:

```text
adds wheel changes
```

Git says:

> "I need to combine both works safely."

---

# Scenario 3 — Different Files Changed

Suppose:

- You changed `app.py`
- Another developer changed `train.py`

Git usually merges automatically.

---

# Example

Before:

```text
app.py
train.py
```

You modify:

```text
app.py
```

Other developer modifies:

```text
train.py
```

Git can combine safely.

No conflict.

---

# Scenario 4 — Same File Changed

Now suppose BOTH developers modify:

```text
app.py
```

---

# Possible Case 1 — Different Lines

You modify:

```python
line 10
```

Other developer modifies:

```python
line 100
```

Git usually merges automatically.

---

# Possible Case 2 — Same Lines

Now BOTH modify:

```python
same lines
```

Git cannot decide.

This creates:

```text
MERGE CONFLICT
```

---

# Example Conflict

Suppose original:

```python
print("hello")
```

You changed:

```python
print("hello from Ali")
```

Other developer changed:

```python
print("hello from Sara")
```

Git cannot know which is correct.

---

# Conflict Markers

Git modifies file like this:

```python
<<<<<<< HEAD
print("hello from Ali")
=======
print("hello from Sara")
>>>>>>> main
```

---

# Meaning

```text
<<<<<<< HEAD
```

your version.

```text
=======
```

separator.

```text
>>>>>>> main
```

incoming version.

---

# What Should You Do?

You must manually edit file.

Example final version:

```python
print("hello from Ali and Sara")
```

Then:

```bash
git add app.py
git commit
```

---

# Visual Diagram

Before merge:

```text
          C (your work)
         /
A --- B
         \
          D (other developer)
```

After merge:

```text
          C
         / \
A --- B ----- M
         \ /
          D
```

`M` = merge commit.

---

# 1. Get Latest Changes

```bash
git pull
```

---

# 2. Safer Pull (Common Modern Workflow)

```bash
git pull --rebase
```

This avoids unnecessary merge commits.

---

# 3. Upload Changes

```bash
git push
```

---

# 4. Check Modified Files

```bash
git status
```

---

# 5. See Differences

```bash
git diff
```

---

# 6. Abort Merge

If merge becomes messy:

```bash
git merge --abort
```

---

# Important Concept — Fast Forward

Suppose GitHub:

```text
A --- B
```

Your local:

```text
A --- B --- C
```

Push is simple because:

```text
history only moves forward
```

called:

```text
fast-forward
```

---

# Non-Fast-Forward Error

Suppose:

GitHub:

```text
A --- B --- D
```

Local:

```text
A --- B --- C
```

Git cannot fast-forward.

Push rejected.

---

# Why Git Rejects Pushes

Git protects repository history.

Without this protection:

- developers could overwrite each other's work
- important commits could disappear

---

# Dangerous Command — Force Push

Sometimes people use:

```bash
git push --force
```

Dangerous.

This can overwrite remote history.

Usually avoid on shared branches.

---

# Analogy

Imagine company archive.

Force push means:

> "Throw away everyone else's recent pages and replace with mine."

Very dangerous.

---

# Better Workflow for Teams

Usually:

```bash
git pull
# resolve conflicts if needed
git push
```

---

# Real-World Team Workflow

---

# Step 1

Get latest changes:

```bash
git pull --rebase
```

---

# Step 2

Modify files.

---

# Step 3

Commit:

```bash
git add .
git commit -m "Added new feature"
```

---

# Step 4

Push:

```bash
git push
```

---

# What If Conflict Happens During Pull?

Git pauses operation.

You resolve conflicts manually.

Then continue:

```bash
git add .
git rebase --continue
```

or if merging:

```bash
git commit
```

---

# Common Branch Strategy

Teams usually avoid everyone working directly on:

```text
main
```

Instead:

```text
main
feature-login
feature-ui
bugfix
```

Each developer works on separate branches.

Later merge into main.

This reduces conflicts.

---

Imagine several engineers editing same machine design.

---

# Safe Case

One edits engine.

Another edits wheels.

No conflict.

---

# Conflict Case

Both edit same screw position.

Now someone must manually decide correct version.

# Q & A

---

## Q1: Can two developers modify same file?

Yes.

Very common.

---

## Q2: Can Git automatically merge same file?

Often yes if different lines changed.

---

## Q3: When does merge conflict happen?

Usually when:

```text
same lines
```

were modified differently.

---

## Q4: Why does push sometimes fail?

Because remote repository changed before your push.

---

## Q5: What does non-fast-forward mean?

Remote history diverged from local history.

---

## Q6: What should I do after push rejection?

Usually:

```bash
git pull --rebase
```

then resolve conflicts if needed.

---

## Q7: Is `git push --force` dangerous?

Yes.

Can overwrite others' work.

---

## Q8: Why use branches in teams?

To isolate work and reduce conflicts.

---

## Q9: Can conflicts happen during pull?

Yes.

Especially during merge/rebase.

---

## Q10: Does Git automatically solve all conflicts?

No.

Human decision is sometimes required.
