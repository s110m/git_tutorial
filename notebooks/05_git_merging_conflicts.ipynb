# Git Merging and Conflict Resolution

---

# Introduction

One of the most important ideas in Git is merging.

Branches allow independent development, but eventually we usually want to combine the work of different branches.

This process is called:

```text
Merging
```

Merging is extremely common in:

- team development
- feature development
- bug fixing
- synchronization with `main`
- collaborative software engineering

---

# What is Merging?

---

# Q: What is merging in Git?

## Answer

Merging means:

> combining the histories and changes of two branches.

Git tries to combine:

- commits
- file changes
- project history

into one unified branch.

---

# Example

Suppose:

```text
A -> B -> C (main)
          \
           D -> E (feature_ai)
```

Now we want to combine `feature_ai` into `main`.

This is merging.

---

# Why Do We Use Merging?

---

# Q: Why do we merge branches?

## Answer

We merge branches because eventually we want:

- new features
- bug fixes
- improvements

to become part of the main project.

---

# Example Scenario

Suppose:

- `main` = stable production code
- `feature_ai` = experimental AI feature

After finishing the AI feature:

```text
feature_ai
```

should become part of:

```text
main
```

So we merge it.

---

# Typical Merge Workflow

---

# Q: How do we merge a branch into `main`?

## Answer

Step 1 — Go to `main`

```bash
git switch main
```

---

Step 2 — Merge branch

```bash
git merge feature_ai
```

Git now tries to combine the histories.

---

# Example Visualization

Before merge:

```text
A -> B -> C (main)
          \
           D -> E (feature_ai)
```

After merge:

```text
A -> B -> C -------- M (main)
          \         /
           D -> E --
```

`M` is the merge commit.

---

# Synchronizing Branches

---

# Q: Do we only merge into `main`?

## Answer

No.

Sometimes we merge branches simply to synchronize development.

---

# Important Real-World Scenario

Suppose:

You are working for two weeks on:

```text
feature_ai
```

Meanwhile:

- another developer fixes bugs
- another feature branch gets merged into `main`
- important updates happen in `main`

Now your branch becomes outdated.

---

# Problem

Your branch may now:

- miss important fixes
- become incompatible
- create bigger future conflicts

---

# Solution

Synchronize your branch with `main`.

---

# Typical Synchronization Workflow

Step 1 — Go to your branch

```bash
git switch feature_ai
```

---

Step 2 — Merge `main`

```bash
git merge main
```

Now your branch receives the latest updates from `main`.

This is extremely common in real projects.

---

# Why Is Early Synchronization Important?

## Answer

Because if branches diverge too much:

- conflicts become larger
- merging becomes harder
- debugging becomes harder

Small frequent merges are usually easier.

---

# Merge Conflicts

---

# Q: What is a merge conflict?

## Answer

A merge conflict happens when Git cannot automatically determine how to combine changes.

Usually this happens when:

> the same part of the same file was modified differently in both branches.

---

# Conflict Visualization

Suppose:

```text
A -> B -> C (main)
          \
           D -> E (feature_ai)
```

Both branches modified:

```python
main.py
```

but differently.

---

# Example

## In `main`

```python
learning_rate = 0.001
```

---

## In `feature_ai`

```python
learning_rate = 0.1
```

Git now has a problem:

```text
Which version is correct?
```

Git cannot decide automatically.

Conflict occurs.

---

# Typical Conflict Error

Example:

```text
CONFLICT (content): Merge conflict in main.py
Automatic merge failed; fix conflicts and then commit the result.
```

---

# Important Idea

Git usually merges automatically if:

- different files changed
- different lines changed

Conflict mainly occurs when:

- same file
- same location
- different modifications

---

# How Git Represents Conflicts

Git inserts conflict markers inside the file.

Example:

```python
<<<<<<< HEAD
learning_rate = 0.001
=======
learning_rate = 0.1
>>>>>>> feature_ai
```

---

# Meaning of Conflict Markers

| Marker | Meaning |
|---|---|
| `<<<<<<< HEAD` | current branch version |
| `=======` | separator |
| `>>>>>>> feature_ai` | incoming branch version |

---

# Resolving Conflicts

---

# Q: How do we resolve conflicts?

## Answer

Conflicts must be resolved manually.

---

# Typical Steps

## Step 1 — Open conflicted files

Git tells which files contain conflicts.

---

## Step 2 — Edit manually

Choose:

- current version
- incoming version
- or combine both

---

# Example

Replace:

```python
<<<<<<< HEAD
learning_rate = 0.001
=======
learning_rate = 0.1
>>>>>>> feature_ai
```

with:

```python
learning_rate = 0.01
```

---

## Step 3 — Stage resolved files

```bash
git add main.py
```

---

## Step 4 — Complete merge commit

```bash
git commit
```

Git creates the merge commit.

---

# Important Idea

Conflict resolution itself becomes a new commit.

Git records:

- how conflict was solved
- the final chosen version

---

# Different Conflict Scenarios

---

# Scenario 1 — Same line modified

Most common conflict.

---

# Scenario 2 — One branch deletes file, another modifies it

Git may ask:

```text
modify/delete conflict
```

You must decide:

- keep file
- delete file

---

# Scenario 3 — Both branches create same file differently

Conflict also occurs.

---

# Merging Into Main Carefully

---

# Q: Why should we be careful merging into `main`?

## Answer

Because `main` is usually:

- stable branch
- production branch
- important branch

Bad merges may break the project.

---

# Common Best Practice

Usually developers:

- create branches from `main`
- work independently
- test carefully
- merge back later

This reduces future problems.

---

# Example Workflow

```text
main
 ├── feature_ai
 ├── feature_web
 └── bugfix_memory
```

Each feature is isolated.

---

# Why Does Conflict Resolution Stay Only in Current Branch?

---

# Q: Suppose we are in `feature_ai` and run:

```bash
git merge main
```

After resolving conflict and committing:

why does `main` itself not change?

## Answer

Because merging modifies:

> the currently active branch only.

---

# Important Rule

When you run:

```bash
git merge another_branch
```

Git says:

> “Bring changes FROM `another_branch`
> INTO the current branch.”

---

# Example

Suppose:

Current branch:

```text
feature_ai
```

Command:

```bash
git merge main
```

Result:

- `feature_ai` changes
- `main` remains untouched

---

# Why Is This Important?

Because branches are independent timelines.

Merging copies history and changes into the current branch only.

---

# Why No Conflict When Merging Back Later?

---

# Q: Suppose:

1. `feature_ai` merges `main`
2. conflict is resolved
3. later `main` merges `feature_ai`

Why often no conflict appears later?

## Answer

Because Git remembers the merge history.

After conflict resolution:

- the merge commit already contains the resolved version
- Git sees that histories are now compatible

So future merges often become automatic.

---

# Visualization

Before synchronization:

```text
A -> B -> C (main)
          \
           D -> E (feature_ai)
```

Conflict occurs.

---

After merge + resolution:

```text
A -> B -> C -------- M (feature_ai)
          \         /
           D -> E --
```

Now histories are connected.

Later merging back to `main` becomes much easier.

---

# Fast-Forward Merge

---

# Q: What is a fast-forward merge?

## Answer

Suppose:

```text
A -> B -> C (main)
          \
           D -> E (feature_ai)
```

and `main` did not change after branch creation.

Then Git may simply move `main` forward:

```text
A -> B -> C -> D -> E (main)
```

No merge commit needed.

This is called:

```text
Fast-forward merge
```

# Final Mental Model

Branches are like:

- parallel timelines
- alternative project histories

Merging is the process of:

> intelligently combining those timelines together.

# Why Conflict Does Not Reappear After Manual Resolution

Suppose we have two branches:

```text
A -> B -> C (main)
          \
           D -> E (feature)
```

Both branches changed the same part of:

```text
main.py
```

So conflict happens.

---

# Step 1 — Merge `main` into `feature`

```bash
git switch feature
git merge main
```

Conflict occurs.

---

# Step 2 — Resolve Conflict Manually

Git shows:

```python
<<<<<<< HEAD
learning_rate = 0.1
=======
learning_rate = 0.001
>>>>>>> main
```

You manually resolve:

```python
learning_rate = 0.01
```

Then:

```bash
git add main.py
git commit
```

---

# Important Result

Git creates a NEW merge commit:

```text
A -> B -> C -------- M (feature)
          \         /
           D -> E --
```

`M` stores:

- both histories
- your manual conflict resolution

This is VERY important.

---

# Step 3 — Later Merge Back Into `main`

Now:

```bash
git switch main
git merge feature
```

Usually NO conflict appears.

Why?

Because Git already sees:

```text
The conflict was resolved previously in merge commit M.
```

Git remembers:

- the merge history
- the resolved version

So it can merge automatically.

---

# Important Idea

Git does NOT only compare files.

Git also compares:

- commit history
- merge history
- common ancestors

This is called:

```text
Three-way merge
```

---

# Final Mental Model

Conflict resolution creates:

```text
a new shared truth
```

inside Git history.

That is why the same conflict usually does not appear again.
