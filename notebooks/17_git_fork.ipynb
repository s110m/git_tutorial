# Understanding Fork in GitHub

A **fork** is one of the most important GitHub concepts, especially in open-source projects.

A fork is:

```text
a complete copy of a repository under your own GitHub account
```

---

# 1. Fork vs Branch

Many beginners confuse:

```text
fork
```

with:

```text
branch
```

But they are different.

---

# Branch

A branch is another development path inside the SAME repository.

```text
github.com/company/project

main
feature-login
bugfix
experiment
```

All branches are inside one repository.

---

# Fork

A fork creates a separate copy of the whole repository.

```text
Original repository:
github.com/company/project

Your fork:
github.com/ali/project
```

Now there are TWO repositories.

---

# Simple Difference

| Concept | Meaning |
|---|---|
| Branch | New line of work inside same repository |
| Fork | New copy of entire repository under another account |

---

# 2. Analogy

Imagine a company building.

---

# Branch

A branch is like:

```text
a different room inside the same building
```

You are still working inside the same company building.

---

# Fork

A fork is like:

```text
copying the whole building and owning your own version
```

You can change your copy freely.

The original building is not changed.

---

# 3. Why Do We Need Fork?

Fork is mainly useful when:

```text
you do NOT have permission to push to the original repository
```

This is common in open-source projects.

---

# Example

Suppose this repository exists:

```text
github.com/pytorch/pytorch
```

You usually cannot push directly to it.

So you:

```text
fork it
```

Then GitHub creates your own copy:

```text
github.com/your-name/pytorch
```

Now you can push to your fork.

---

# 4. Fork Workflow Overview

Typical open-source workflow:

```text
1. Fork original repository
2. Clone your fork
3. Create a branch
4. Make changes
5. Commit changes
6. Push branch to your fork
7. Open Pull Request to original repository
```

# 6. Step-by-Step Fork Workflow

---

# Step 1 — Fork the Repository on GitHub

Go to the original repository page.

Click:

```text
Fork
```

GitHub creates a copy under your account.

Before:

```text
github.com/company/project
```

After fork:

```text
github.com/ali/project
```

---

# Step 2 — Clone Your Fork

Now clone YOUR fork, not the original repository.

```bash
git clone https://github.com/ali/project.git
cd project
```

After cloning:

```text
Your GitHub fork
        |
        | clone
        v
Your local computer
```

---

# Step 3 — Check Remote

Run:

```bash
git remote -v
```

You may see:

```text
origin  https://github.com/ali/project.git (fetch)
origin  https://github.com/ali/project.git (push)
```

Here:

```text
origin = your fork
```

not the original repository.

---

# 7. What Is `origin` Here?

After cloning your fork:

```text
origin
```

points to your fork.

```text
origin -> github.com/ali/project
```

Diagram:

```text
Your Laptop
     |
     | origin
     v
Your Fork
```

---

# 8. Add Original Repository as `upstream`

Usually we also add the original repository as:

```text
upstream
```

Command:

```bash
git remote add upstream https://github.com/company/project.git
```

Now check:

```bash
git remote -v
```

You may see:

```text
origin    https://github.com/ali/project.git (fetch)
origin    https://github.com/ali/project.git (push)
upstream  https://github.com/company/project.git (fetch)
upstream  https://github.com/company/project.git (push)
```

Usually:

| Remote | Meaning |
|---|---|
| `origin` | Your fork |
| `upstream` | Original repository |

---

# 9. Why Do We Need `upstream`?

Because the original project may continue changing.

Other developers may add commits to:

```text
github.com/company/project
```

You need a way to update your fork/local project from the original repository.

---

# 10. Sync Your Fork with Original Repository

To get latest changes from original repository:

```bash
git fetch upstream
```

Then update your local `main`:

```bash
git switch main
git merge upstream/main
```

or:

```bash
git switch main
git rebase upstream/main
```

Then update your GitHub fork:

```bash
git push origin main
```

---

# Diagram: Syncing Fork

```text
Original Repository
upstream/main
        |
        | fetch/merge
        v
Your Local main
        |
        | push
        v
Your Fork
origin/main
```

---

# 11. Create a Branch in Your Fork

Do not usually work directly on `main`.

Create a feature branch:

```bash
git switch -c fix-bug
```

Diagram:

```text
main:
A --- B --- C

fix-bug:
A --- B --- C
```

After changes:

```text
main:
A --- B --- C

fix-bug:
A --- B --- C --- D
```

---

# 12. Make Changes and Commit

Edit files.

Then:

```bash
git status
git add .
git commit -m "Fix bug in login page"
```

Now your branch has your work.

---

# 13. Push Branch to Your Fork

```bash
git push -u origin fix-bug
```

Important:

```text
origin = your fork
```

So you are pushing to:

```text
github.com/ali/project
```

not directly to original repository.

---

# Diagram

```text
Your Laptop branch
fix-bug
        |
        | push
        v
Your Fork branch
origin/fix-bug
```

---

# 14. Open Pull Request

Now go to GitHub.

Open a Pull Request from:

```text
ali/project:fix-bug
```

into:

```text
company/project:main
```

Diagram:

```text
Your Fork / fix-bug
        |
        | Pull Request
        v
Original Repository / main
```

---

# Important Direction

Pull Request direction is usually:

```text
your fork branch  --->  original repository main
```

Not the opposite.

---

# 15. What Happens After Pull Request?

The original project maintainers can:

- review your code
- comment
- request changes
- run tests
- approve
- merge your changes

Your fork remains separate.

The original repository changes only if maintainers merge your PR.

---

# 16. If Maintainer Requests Changes

You do not need to create a new Pull Request.

Just continue on the same branch:

```bash
git switch fix-bug
```

Make changes.

Then:

```bash
git add .
git commit -m "Address review comments"
git push
```

The Pull Request updates automatically.

---

# 17. Fork vs Clone

Very important difference.

---

# Fork

GitHub operation:

```text
creates a copy on GitHub
```

---

# Clone

Git operation:

```text
downloads repository to your computer
```

---

# Diagram

```text
Fork:
GitHub original -> GitHub your copy

Clone:
GitHub repository -> your laptop
```

---

# 18. Fork vs Pull Request

Fork is:

```text
your copy of repository
```

Pull Request is:

```text
request to merge your changes into another repository or branch
```

---

# 19. Fork vs Branch

| Feature | Branch | Fork |
|---|---|---|
| Location | Same repository | Separate repository |
| Permission needed | Usually write access | No write access needed |
| Common use | Team internal work | Open-source contribution |
| Can have branches? | Yes | A fork can also have branches |
| Pull request possible? | Yes | Yes |

---

# 20. Common Professional/Open-Source Workflow

```bash
# 1. Clone your fork
git clone https://github.com/ali/project.git
cd project

# 2. Add original repo as upstream
git remote add upstream https://github.com/company/project.git

# 3. Update from original
git fetch upstream
git switch main
git merge upstream/main

# 4. Create branch
git switch -c fix-bug

# 5. Make changes
git add .
git commit -m "Fix bug"

# 6. Push to your fork
git push -u origin fix-bug
```

Then on GitHub:

```text
Open Pull Request:
ali/project:fix-bug -> company/project:main
```

---

# 21. When Do We Use Fork?

Use fork when:

- you do not have write permission
- you contribute to open-source project
- you want your own independent copy
- you want to experiment freely
- you want to propose changes without touching original repo

---

# 22. When Do We Use Branch Instead?

Use branch when:

- you are inside a company/team repository
- you have write access
- project owner allows branch pushing
- you work on normal features/bug fixes

---

# 23. Can Fork Become Outdated?

Yes.

Original repository may receive new commits.

Example:

```text
Original:
A --- B --- C --- D --- E

Your fork:
A --- B --- C
```

Your fork is behind.

You need to sync with upstream.

---

# 24. Updating an Outdated Fork

```bash
git fetch upstream
git switch main
git merge upstream/main
git push origin main
```

Now:

```text
Your fork becomes updated
```

---

# 25. Can I Delete a Fork?

Yes.

Since it is your repository, you can delete it from GitHub settings.

This does not delete the original repository.

---

# 26. Analogy of Engineers

Imagine a famous company has an official machine blueprint.

You are not allowed to directly edit it.

So you:

1. copy the full blueprint book
2. edit your copy
3. send your edited pages to the company
4. company engineers review them
5. if they like them, they add changes to official blueprint

That full copy is:

```text
fork
```

```text
pull request
```

# 27. Q & A

---

## Q1: Is fork the same as branch?

No.

A fork is a separate repository.

A branch is inside one repository.

---

## Q2: Is fork local or on GitHub?

Fork is on GitHub.

Clone is local.

---

## Q3: Do I need permission to fork?

Usually no, for public repositories.

---

## Q4: Why do open-source projects use forks?

Because random contributors usually cannot push directly to original repository.

---

## Q5: After fork, where do I push?

Usually to your fork:

```bash
git push origin branch-name
```

---

## Q6: What is `origin` after cloning my fork?

`origin` usually points to your fork.

---

## Q7: What is `upstream`?

`upstream` usually points to original repository.

---

## Q8: Do I need upstream?

Not always, but it is very useful to sync with original project.

---

## Q9: Can my fork have many branches?

Yes.

A fork is a full repository, so it can have many branches.

---

## Q10: Does fork change the original repository?

No.

Original repository changes only if maintainers merge your Pull Request.

---

## Q11: Can I make Pull Request from fork?

Yes.

That is one of the main purposes of fork.

---

## Q12: Is fork only for open-source?

Mostly common there, but it can also be used in organizations.

---

## Q13: Can my fork become behind original?

Yes.

You should sync it with upstream.

---

## Q14: Can I delete my fork safely?

Yes.

It does not delete the original repository.
