# Connecting a Local Git Repository to GitHub

Suppose you already have:

- a local Git repository
- several branches
- commits in your local machine

and now you want to push everything into GitHub.

This guide explains the common commands and their meanings.

---

# Big Picture

You have:

```text
Your Computer (local git repo)
```

and want to connect it to:

```text
GitHub repository (remote repo)
```

---

# Step 1 — Create a GitHub Repository

Go to:

```text
https://github.com
```

Create:

```text
New Repository
```

Example:

```text
testing
```

GitHub gives you a URL like:

```text
https://github.com/s110m/testing.git
```

This URL is the address of your remote repository.

---

# Analogy

Imagine:

- Local repository = your notebook at home
- GitHub repository = cloud backup server

You now want Git to know:

> "Where is my cloud backup?"

---

# Step 2 — Rename Branch to `main`

Command:

```bash
git branch -M main
```

---

# What Does It Mean?

Historically Git used:

```text
master
```

as default branch name.

Modern GitHub usually prefers:

```text
main
```

This command means:

```text
Rename current branch to main
```

---

# What Does `-M` Mean?

`-M` means:

```text
force rename
```

even if branch already exists.

---

# Example

Before:

```text
master
```

After:

```text
main
```

Diagram:

```text
Before:
HEAD -> master

After:
HEAD -> main
```

---

# Step 3 — Add Remote Repository

Command:

```bash
git remote add origin https://github.com/s110m/testing.git
```

---

# What Does This Mean?

This command says:

> "From now on, the name `origin` refers to this GitHub repository URL."

---

# Important Clarification

`origin` is NOT GitHub itself.

It is simply a nickname.

You could write:

```bash
git remote add banana https://github.com/s110m/testing.git
```

and it would still work.

But conventionally we use:

```text
origin
```

---

# Analogy

Imagine:

```text
origin = nickname in your phone contacts
```

Instead of writing full phone number every time.

---

# Visual Diagram

After this command:

```text
Local Repository
      |
      |  origin
      ↓
https://github.com/s110m/testing.git
```

---

# How to See Remotes

Command:

```bash
git remote -v
```

Example output:

```text
origin  https://github.com/s110m/testing.git (fetch)
origin  https://github.com/s110m/testing.git (push)
```

---

# Step 4 — Push Branch to GitHub

Command:

```bash
git push -u origin main
```

---

# Meaning of This Command

This means:

> "Push my local `main` branch to remote `origin`."

---

# Breaking It Down

---

## `git push`

Means:

```text
Send commits to remote repository
```

---

## `origin`

Means:

```text
the remote repository nickname
```

---

## `main`

Means:

```text
push local branch main
```

---

# What Does `-u` Mean?

`-u` means:

```text
--set-upstream
```

This creates a tracking relationship.

After this:

```bash
git push
```

and:

```bash
git pull
```

automatically know:

```text
which remote branch to use
```

---

# Analogy

Without `-u`:

Git asks:

> "Which remote branch should I use?"

With `-u`:

Git remembers:

> "main tracks origin/main"
```

---

# Diagram

Before push:

```text
Local:
main -> D

GitHub:
(empty)
```

After push:

```text
Local:
main -> D

GitHub:
origin/main -> D
```

---

# Important Question

# Does This Push ALL Branches?

## NO.

This command:

```bash
git push -u origin main
```

ONLY pushes:

```text
main branch
```

---

# Example

Suppose local repository has:

```text
main
feature-login
experiment
bugfix
```

After:

```bash
git push -u origin main
```

GitHub only receives:

```text
main
```

---

# How to Push Other Branches

You must push each branch separately.

Example:

```bash
git push -u origin feature-login
```

or:

```bash
git push -u origin experiment
```

---

# Diagram

Local:

```text
main
feature
experiment
```

GitHub after pushing only main:

```text
origin/main
```

GitHub after pushing feature:

```text
origin/main
origin/feature
```

---

# Why Doesn't Git Push Everything Automatically?

Because:

- some branches are experimental
- some are temporary
- some are private work

Git avoids accidentally uploading everything.

---

# Common Workflow

---

# Push Main Branch

```bash
git push -u origin main
```

---

# Push Feature Branch

```bash
git push -u origin feature
```

---

# Later Pushes Become Simpler

After upstream tracking is set:

```bash
git push
```

is enough.

Git automatically knows destination.

---

# Tracking Relationship

After:

```bash
git push -u origin main
```

Git creates:

```text
local main  <---->  origin/main
```

---

# What is `origin/main`?

Very important concept.

---

# Difference

| Name | Meaning |
|---|---|
| `main` | Your local branch |
| `origin/main` | Remote-tracking branch |

---

# Diagram

```text
Your Computer:
main

GitHub:
origin/main
```

---

# Fetching Updates

Suppose someone updates GitHub repository.

You can retrieve latest remote information using:

```bash
git fetch
```

This updates:

```text
origin/main
```

but NOT your local branch yet.

# Q & A

---

## Q1: Does `git push` push all branches automatically?

No.

Only specified branch is pushed.

---

## Q2: How do I push another branch?

Example:

```bash
git push -u origin feature
```

---

## Q3: What is `origin`?

Just a nickname for remote repository URL.

---

## Q4: Can I use another name instead of origin?

Yes.

Example:

```bash
git remote add cloud https://github.com/...
```

But `origin` is standard convention.

---

## Q5: What does `-u` do?

It sets upstream tracking.

After that:

```bash
git push
```

works automatically.

---

## Q6: What is `origin/main`?

It is Git's local knowledge of remote branch.

---

## Q7: Does `git remote add` upload anything?

No.

It only saves remote address.

---

## Q8: What if branch already exists on GitHub?

Git tries to merge histories if compatible.

Sometimes you may need:

```bash
git pull
```

before pushing.

---

## Q9: Can GitHub repository contain many branches?

Yes.

Very common in team development.

---

## Q10: How do I see all local branches?

```bash
git branch
```

---

## Q11: How do I see remote branches?

```bash
git branch -r
```
